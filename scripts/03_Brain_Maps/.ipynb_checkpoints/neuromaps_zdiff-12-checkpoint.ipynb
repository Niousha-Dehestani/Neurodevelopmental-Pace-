```python
# Neuromaps import
from neuromaps import datasets, transforms
import nibabel as nib
from nibabel.freesurfer import io
import os
import numpy as np
import pandas as pd
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import matplotlib as mpl
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
from kneed import KneeLocator
from matplotlib.ticker import FormatStrFormatter
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

from neuromaps import datasets, transforms
import nibabel as nib
import os


url = "https://raw.githubusercontent.com/richardajdear/AHBA_gradients/ab7939cf1811cba6296b882e35e60b09fed7d653/outputs/ahba_dme_hcp_top8kgenes_scores.csv"
df = pd.read_csv(url)
# load HCP MMP1.0 parcellaiton
atlas = nib.load("/Users/nioushad/Downloads/Q1-Q6_RelatedValidation210.CorticalAreas_dil_Final_Final_Areas_Group_Colors.32k_fs_LR.dlabel.nii")
atlas_data = atlas.get_fdata()[0]
atlas_label_dict = dict(list(atlas.header.get_index_map(0).named_maps)[0].label_table)
atlas_label_names = {i:atlas_label_dict[i].label for i in atlas_label_dict}
atlas_label_indices = {atlas_label_dict[i].label:i for i in atlas_label_dict}


# 1. Initialize Vertex Maps (Glasser-space)
# Using your exact logic for bilateral projection
Gene_PC1_vert = atlas_data * np.nan
Gene_PC2_vert = atlas_data * np.nan
Gene_PC3_vert = atlas_data * np.nan

for i, row in df.iterrows():
    # Symmetric Projection (L + R)
    l_mask = (atlas_data == row.id)
    r_mask = (atlas_data == (row.id + 180))
    combined = l_mask | r_mask
    
    Gene_PC1_vert[combined] = row.C1
    Gene_PC2_vert[combined] = row.C2
    Gene_PC3_vert[combined] = row.C3

# 2. Load the DK Target Atlas
dk_atlas = nib.load("/Users/nioushad/Documents/Doc_p/fsaverage.aparc.32k_fs_LR.dlabel.nii")
dk_data = dk_atlas.get_fdata()[0]
dk_labels = list(dk_atlas.header.get_index_map(0).named_maps)[0].label_table

# 3. Aggregate Vertices into DK Parcels
dk_results = []

for idx, label_obj in dk_labels.items():
    label_name = label_obj.label
    
    # Filter out non-cortical areas
    if label_name in ['???', 'unknown', 'Medial_Wall', 'Background']:
        continue
    
    # Create mask for current DK ROI
    dk_mask = (dk_data == idx)
    
    if np.any(dk_mask):
        dk_results.append({
            'DK_ROI': label_name,
            'PC1': np.nanmean(Gene_PC1_vert[dk_mask]),
            'PC2': np.nanmean(Gene_PC2_vert[dk_mask]),
            'PC3': np.nanmean(Gene_PC3_vert[dk_mask])
        })

# 4. Final Dataframe
df_dk_mapped = pd.DataFrame(dk_results)
print("✅ Successfully parcellated Gene PCs into DK atlas.")
print(df_dk_mapped.head())


df_dk_mapped = df_dk_mapped.rename(columns={'DK_ROI':'ROI'})
vol_mapping = {
    'L_bankssts': 'L.bankssts',
    'L_caudalanteriorcingulate': 'L.caudalanteriorcingulate',
    'L_caudalmiddlefrontal': 'L.caudalmiddlefrontal',
    'L_cuneus': 'L.cuneus',
    'L_entorhinal': 'L.entorhinal',
    'L_fusiform': 'L.fusiform',
    'L_inferiorparietal': 'L.inferiorparietal',
    'L_inferiortemporal': 'L.inferiortemporal',
    'L_isthmuscingulate': 'L.isthmuscingulate',
    'L_lateraloccipital':'L.lateraloccipital',
    'L_lateralorbitofrontal': 'L.lateralorbitofrontal',
    'L_lingual':'L.lingual',
    'L_medialorbitofrontal': 'L.medialorbitofrontal',
    'L_middletemporal': 'L.middletemporal',
    'L_parahippocampal': 'L.parahippocampal',
    'L_paracentral': 'L.paracentral', 
    'L_parsopercularis':'L.parsopercularis',
    'L_parsorbitalis': 'L.parsorbitalis',
    'L_parstriangularis': 'L.parstriangularis',  
    'L_pericalcarine': 'L.pericalcarine',  
    'L_postcentral': 'L.postcentral',
    'L_posteriorcingulate': 'L.posteriorcingulate',
    'L_precentral': 'L.precentral',  
    'L_precuneus': 'L.precuneus',   
    'L_rostralanteriorcingulate': 'L.rostralanteriorcingulate',
    'L_rostralmiddlefrontal': 'L.rostralmiddlefrontal',
    'L_superiorfrontal': 'L.superiorfrontal',
    'L_superiorparietal': 'L.superiorparietal',
    'L_superiortemporal': 'L.superiortemporal',
    'L_supramarginal': 'L.supramarginal',  
    'L_frontalpole': 'L.frontalpole',
    'L_temporalpole': 'L.temporalpole',
    'L_transversetemporal' : 'L.transversetemporal',
    'L_insula': 'L.insula',

   'R_bankssts': 'R.bankssts',
    'R_caudalanteriorcingulate': 'R.caudalanteriorcingulate',
    'R_caudalmiddlefrontal': 'R.caudalmiddlefrontal',
    'R_cuneus': 'R.cuneus',
    'R_entorhinal': 'R.entorhinal',
    'R_fusiform': 'R.fusiform',
    'R_inferiorparietal': 'R.inferiorparietal',
    'R_inferiortemporal': 'R.inferiortemporal',
    'R_isthmuscingulate': 'R.isthmuscingulate',
    'R_lateraloccipital':'R.lateraloccipital',
    'R_lateralorbitofrontal': 'R.lateralorbitofrontal',
    'R_lingual':'R.lingual',
    'R_medialorbitofrontal': 'R.medialorbitofrontal',
    'R_middletemporal': 'R.middletemporal',
    'R_parahippocampal': 'R.parahippocampal',
    'R_paracentral': 'R.paracentral', 
    'R_parsopercularis':'R.parsopercularis',
    'R_parsorbitalis': 'R.parsorbitalis',
    'R_parstriangularis': 'R.parstriangularis',  
    'R_pericalcarine': 'R.pericalcarine',  
    'R_postcentral': 'R.postcentral',
    'R_posteriorcingulate': 'R.posteriorcingulate',
    'R_precentral': 'R.precentral',  
    'R_precuneus': 'R.precuneus',   
    'R_rostralanteriorcingulate': 'R.rostralanteriorcingulate',
    'R_rostralmiddlefrontal': 'R.rostralmiddlefrontal',
    'R_superiorfrontal': 'R.superiorfrontal',
    'R_superiorparietal': 'R.superiorparietal',
    'R_superiortemporal': 'R.superiortemporal',
    'R_supramarginal': 'R.supramarginal',  
    'R_frontalpole': 'R.frontalpole',
    'R_temporalpole': 'R.temporalpole',
    'R_transversetemporal' : 'R.transversetemporal',
    'R_insula': 'R.insula',}

df_dk_mapped['ROI'] = df_dk_mapped['ROI'].replace(vol_mapping)
df_dk_mapped.to_csv(
    '/Users/nioushad/Documents/Doc_p/myJupyter/normative modeling/code/component/release6/hpc/df_dk_mapped.csv')



zdiff = pd.read_csv(
    '/Users/nioushad/Documents/Doc_p/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/processt1/Results_Longitudinal/PC1_Loadings_V1_V2_healthy.csv'
)
zdiff = zdiff.rename(columns={'Unnamed: 0': 'ROI'})

vol_mapping = {'mr_y_smri__vol__aseg__ag__lh_sum': 'AMYGDALA_LEFT',
                 'mr_y_smri__vol__aseg__ag__rh_sum': 'AMYGDALA_RIGHT',
                 'mr_y_smri__vol__aseg__cd__lh_sum': 'CAUDATE_LEFT',
                 'mr_y_smri__vol__aseg__cd__rh_sum': 'CAUDATE_RIGHT',
                 'mr_y_smri__vol__aseg__hc__lh_sum': 'HIPPOCAMPUS_LEFT',
                 'mr_y_smri__vol__aseg__hc__rh_sum': 'HIPPOCAMPUS_RIGHT',
                 'mr_y_smri__vol__aseg__pl__lh_sum':'PALLIDUM_LEFT', 
                 'mr_y_smri__vol__aseg__pl__rh_sum': 'PALLIDUM_RIGHT',
                 'mr_y_smri__vol__aseg__pt__lh_sum': 'PUTAMEN_LEFT',
                 'mr_y_smri__vol__aseg__pt__rh_sum': 'PUTAMEN_RIGHT',
                 'mr_y_smri__vol__aseg__th__lh_sum': 'THALAMUS_LEFT',
                 'mr_y_smri__vol__aseg__th__rh_sum': 'THALAMUS_RIGHT',
                 'mr_y_smri__vol__aseg__ab__lh_sum': 'ACCUMBENS_LEFT',
                 'mr_y_smri__vol__aseg__ab__rh_sum': 'ACCUMBENS_RIGHT',
                'mr_y_smri__vol__dsk__bstmps__lh_sum': 'L.bankssts',
                'mr_y_smri__vol__dsk__cac__lh_sum': 'L.caudalanteriorcingulate',
                'mr_y_smri__vol__dsk__cmfrt__lh_sum': 'L.caudalmiddlefrontal',
                'mr_y_smri__vol__dsk__cn__lh_sum': 'L.cuneus',
                'mr_y_smri__vol__dsk__er__lh_sum': 'L.entorhinal',
                'mr_y_smri__vol__dsk__ff__lh_sum': 'L.fusiform',
                'mr_y_smri__vol__dsk__iprt__lh_sum': 'L.inferiorparietal',
                'mr_y_smri__vol__dsk__itmp__lh_sum': 'L.inferiortemporal',
                'mr_y_smri__vol__dsk__ic__lh_sum': 'L.isthmuscingulate',
                'mr_y_smri__vol__dsk__locc__lh_sum':'L.lateraloccipital',
                'mr_y_smri__vol__dsk__lobfrt__lh_sum': 'L.lateralorbitofrontal',
                'mr_y_smri__vol__dsk__lg__lh_sum':'L.lingual',
                'mr_y_smri__vol__dsk__mobfrt__lh_sum': 'L.medialorbitofrontal',
                'mr_y_smri__vol__dsk__mtmp__lh_sum': 'L.middletemporal',
                'mr_y_smri__vol__dsk__ph__lh_sum': 'L.parahippocampal',
                'mr_y_smri__vol__dsk__pactr__lh_sum': 'L.paracentral', 
                'mr_y_smri__vol__dsk__pop__lh_sum':'L.parsopercularis',
                'mr_y_smri__vol__dsk__pob__lh_sum': 'L.parsorbitalis',
                'mr_y_smri__vol__dsk__ptg__lh_sum': 'L.parstriangularis',  
                'mr_y_smri__vol__dsk__pcc__lh_sum': 'L.pericalcarine',  
                'mr_y_smri__vol__dsk__poctr__lh_sum': 'L.postcentral',
                'mr_y_smri__vol__dsk__pcg__lh_sum': 'L.posteriorcingulate',
                'mr_y_smri__vol__dsk__prctr__lh_sum': 'L.precentral',  
                'mr_y_smri__vol__dsk__prcn__lh_sum': 'L.precuneus',   
                'mr_y_smri__vol__dsk__rac__lh_sum': 'L.rostralanteriorcingulate',
                'mr_y_smri__vol__dsk__rmfrt__lh_sum': 'L.rostralmiddlefrontal',
                'mr_y_smri__vol__dsk__sfrt__lh_sum': 'L.superiorfrontal',
                'mr_y_smri__vol__dsk__sprt__lh_sum': 'L.superiorparietal',
                'mr_y_smri__vol__dsk__stmp__lh_sum': 'L.superiortemporal',
                'mr_y_smri__vol__dsk__sm__lh_sum': 'L.supramarginal',  
                'mr_y_smri__vol__dsk__pfrt__lh_sum': 'L.frontalpole',
                'mr_y_smri__vol__dsk__ptmp__lh_sum': 'L.temporalpole',
                'mr_y_smri__vol__dsk__ttmp__lh_sum' : 'L.transversetemporal',
                'mr_y_smri__vol__dsk__ins__lh_sum': 'L.insula',
            
               'mr_y_smri__vol__dsk__bstmps__rh_sum': 'R.bankssts',
                'mr_y_smri__vol__dsk__cac__rh_sum': 'R.caudalanteriorcingulate',
                'mr_y_smri__vol__dsk__cmfrt__rh_sum': 'R.caudalmiddlefrontal',
                'mr_y_smri__vol__dsk__cn__rh_sum': 'R.cuneus',
                'mr_y_smri__vol__dsk__er__rh_sum': 'R.entorhinal',
                'mr_y_smri__vol__dsk__ff__rh_sum': 'R.fusiform',
                'mr_y_smri__vol__dsk__iprt__rh_sum': 'R.inferiorparietal',
                'mr_y_smri__vol__dsk__itmp__rh_sum': 'R.inferiortemporal',
                'mr_y_smri__vol__dsk__ic__rh_sum': 'R.isthmuscingulate',
                'mr_y_smri__vol__dsk__locc__rh_sum':'R.lateraloccipital',
                'mr_y_smri__vol__dsk__lobfrt__rh_sum': 'R.lateralorbitofrontal',
                'mr_y_smri__vol__dsk__lg__rh_sum':'R.lingual',
                'mr_y_smri__vol__dsk__mobfrt__rh_sum': 'R.medialorbitofrontal',
                'mr_y_smri__vol__dsk__mtmp__rh_sum': 'R.middletemporal',
                'mr_y_smri__vol__dsk__ph__rh_sum': 'R.parahippocampal',
                'mr_y_smri__vol__dsk__pactr__rh_sum': 'R.paracentral', 
                'mr_y_smri__vol__dsk__pop__rh_sum':'R.parsopercularis',
                'mr_y_smri__vol__dsk__pob__rh_sum': 'R.parsorbitalis',
                'mr_y_smri__vol__dsk__ptg__rh_sum': 'R.parstriangularis',  
                'mr_y_smri__vol__dsk__pcc__rh_sum': 'R.pericalcarine',  
                'mr_y_smri__vol__dsk__poctr__rh_sum': 'R.postcentral',
                'mr_y_smri__vol__dsk__pcg__rh_sum': 'R.posteriorcingulate',
                'mr_y_smri__vol__dsk__prctr__rh_sum': 'R.precentral',  
                'mr_y_smri__vol__dsk__prcn__rh_sum': 'R.precuneus',   
                'mr_y_smri__vol__dsk__rac__rh_sum': 'R.rostralanteriorcingulate',
                'mr_y_smri__vol__dsk__rmfrt__rh_sum': 'R.rostralmiddlefrontal',
                'mr_y_smri__vol__dsk__sfrt__rh_sum': 'R.superiorfrontal',
                'mr_y_smri__vol__dsk__sprt__rh_sum': 'R.superiorparietal',
                'mr_y_smri__vol__dsk__stmp__rh_sum': 'R.superiortemporal',
                'mr_y_smri__vol__dsk__sm__rh_sum': 'R.supramarginal',  
                'mr_y_smri__vol__dsk__pfrt__rh_sum': 'R.frontalpole',
                'mr_y_smri__vol__dsk__ptmp__rh_sum': 'R.temporalpole',
                'mr_y_smri__vol__dsk__ttmp__rh_sum' : 'R.transversetemporal',
                'mr_y_smri__vol__dsk__ins__rh_sum': 'R.insula',}



zdiff['ROI'] = zdiff['ROI'].replace(vol_mapping)

# Create a list of ROIs that are present in both dataframes
# This usually results in the 68 cortical regions (removing subcortical)
common_rois = [roi for roi in df_dk_mapped['ROI'].values if roi in zdiff['ROI'].values]

print(f"Number of matching regions: {len(common_rois)}")

# Set ROI as the index so we can select by name
df1_temp = zdiff.set_index('ROI')
df2_temp = df_dk_mapped.set_index('ROI')

# Filter and Reorder both dataframes to match 'common_rois'
df1_aligned = df1_temp.loc[common_rois]
df2_aligned = df2_temp.loc[common_rois]

# Reset the index if you want 'ROI' back as a column
df1_final = df1_aligned.reset_index()
df2_final = df2_aligned.reset_index()



########################################################################
# all available maps
len(datasets.available_annotations())
########################################################################

########################################################################
# The following fixes an issue:
# Nilearn Monkey patch
from pathlib import Path
import neuromaps.datasets.annotations as annots

# Keep reference to the original _fetch_file
orig_fetch_file = annots._fetch_file

def patched_fetch_file(url, data_dir, *args, **kwargs):
    """Ensure data_dir is always a Path."""
    if isinstance(data_dir, str):
        data_dir = Path(data_dir)
    return orig_fetch_file(url, data_dir, *args, **kwargs)



# Apply the patch
annots._fetch_file = patched_fetch_file
########################################################################

########################################################################
# this downloads all maps
all_annotations = datasets.fetch_annotation(source="all")
########################################################################

########################################################################
# This removes some duplicates
# remove duplicate volumetric maps of the beliveau2017 source
all_annotations = {
    k: v for k, v in all_annotations.items()
    if not (k[0] == "beliveau2017" and "MNI" in k[2])
}
# remove duplicate volumetric maps of the norgaard2021 source
all_annotations = {
    k: v for k, v in all_annotations.items()
    if not (k[0] == "norgaard2021" and "MNI" in k[2])
}
# remove neuromaps missing both hemispheres data
all_annotations = {
    k: v for k, v in all_annotations.items()
    if not (k[2] == "fsLR" and isinstance(v, str))
}


# --- paths (you already downloaded these) ---
label_dir = os.path.expanduser('~/freesurfer/subjects/fsaverage5/label')

# Desikan–Killiany annotations (left/right)
dk_l, ctab_l, names_l = nib.freesurfer.read_annot(os.path.join(label_dir, 'lh.aparc.annot'))
dk_r, ctab_r, names_r = nib.freesurfer.read_annot(os.path.join(label_dir, 'rh.aparc.annot'))



# you can load the DK atlas like so: (from freesurfer directory)
dk_l = nib.freesurfer.read_annot("/Users/nioushad/freesurfer/subjects/fsaverage5/label/lh.aparc.annot")[0]

dk_r = nib.freesurfer.read_annot("/Users/nioushad/freesurfer/subjects/fsaverage5/label/rh.aparc.annot")[0]

########################################################################

# Decode names from bytes -> str
names_l = [n.decode() if isinstance(n, bytes) else n for n in names_l]
names_r = [n.decode() if isinstance(n, bytes) else n for n in names_r]



def parcellate_hemi(image, labels, names, hemi):
    """Compute mean per DK parcel for one hemisphere, skipping empty/NaN-only parcels."""
    out = []
    for lab in np.unique(labels):
        if lab < 0:
            continue
        name = names[lab] if lab < len(names) else f"label_{lab}"
        if name.lower() in ("unknown", "???"):
            continue

        mask = labels == lab
        if mask.sum() == 0:
            continue

        vals = image[mask]
        # keep only finite values
        vals = vals[np.isfinite(vals)]
        if vals.size == 0:  # all NaN → skip to avoid warnings
            continue

        out.append({
            "hemi": hemi,
            "label_index": int(lab),
            "region": name,
            "n_verts": int(vals.size),
            "mean": float(np.mean(vals)),
            "std": float(np.std(vals, ddof=1)) if vals.size > 1 else 0.0,
        })
    return pd.DataFrame(out)



# === Bilateral BrainSMASH: PC1 loadings (vol_lh + vol_rh) vs ABAGEN (dk_parcalslh + dk_parcalsrh) ===
from nilearn import datasets, surface
import numpy as np
import pandas as pd
from brainsmash.mapgen.base import Base

# 0) fsaverage5 meshes
fsavg = datasets.fetch_surf_fsaverage(mesh='fsaverage5')
lh_mesh = surface.load_surf_mesh(fsavg['pial_left'])
rh_mesh = surface.load_surf_mesh(fsavg['pial_right'])
coords_lh = np.asarray(lh_mesh.coordinates)
coords_rh = np.asarray(rh_mesh.coordinates)



# === Bilateral BrainSMASH: PC1 loadings (vol_lh + vol_rh) vs ABAGEN (dk_parcalslh + dk_parcalsrh) ===
from nilearn import datasets, surface
import numpy as np
import pandas as pd
from brainsmash.mapgen.base import Base

# 0) fsaverage5 meshes
fsavg = datasets.fetch_surf_fsaverage(mesh='fsaverage5')
lh_mesh = surface.load_surf_mesh(fsavg['pial_left'])
rh_mesh = surface.load_surf_mesh(fsavg['pial_right'])
coords_lh = np.asarray(lh_mesh.coordinates)
coords_rh = np.asarray(rh_mesh.coordinates)

# 1) sanity: label arrays must match vertex counts
assert dk_l.shape[0] == coords_lh.shape[0], "LH annot != LH surface vertex count"
assert dk_r.shape[0] == coords_rh.shape[0], "RH annot != RH surface vertex count"

# 2) name -> label index maps from annot names (already decoded str)
name_map_l = {name: i for i, name in enumerate(names_l)}
name_map_r = {name: i for i, name in enumerate(names_r)}




# 1. Ensure the Region names in zdiff_lh are cleaned (as we did previously)
zdiff['Region'] = zdiff['ROI'].str.replace('L.', '', regex=False).str.replace('R.', '', regex=False)

# 2. Create a mapping dictionary from ROI to Region
roi_to_region = dict(zip(zdiff['ROI'], zdiff['Region']))

# 3. Map this dictionary onto the df_dk_mapped dataframe
df_dk_mapped['Region'] = df_dk_mapped['ROI'].map(roi_to_region)


# Add hemisphere column
zdiff['hemi'] = np.where(
    zdiff['ROI'].str.contains('L.'), 'lh',
    np.where(zdiff['ROI'].str.contains('R.'), 'rh', 'unknown')
)
zdiff_lh = zdiff[zdiff['hemi'] == 'lh']
zdiff_rh = zdiff[zdiff['hemi'] == 'rh']



df_dk_mapped['hemi'] = np.where(
    df_dk_mapped['ROI'].str.contains('L.'), 'lh',
    np.where(df_dk_mapped['ROI'].str.contains('R.'), 'rh', 'unknown')
)
df_dk_mapped_lh = df_dk_mapped[df_dk_mapped['hemi'] == 'lh']
df_dk_mapped_rh = df_dk_mapped[df_dk_mapped['hemi'] == 'rh']



DK_ORDER = [
 'bankssts','caudalanteriorcingulate','caudalmiddlefrontal','cuneus','entorhinal',
 'frontalpole','fusiform','inferiorparietal','inferiortemporal','insula',
 'isthmuscingulate','lateraloccipital','lateralorbitofrontal','lingual',
 'medialorbitofrontal','middletemporal','paracentral','parahippocampal',
 'parsopercularis','parsorbitalis','parstriangularis','pericalcarine',
 'postcentral','posteriorcingulate','precentral','precuneus',
 'rostralanteriorcingulate','rostralmiddlefrontal','superiorfrontal',
 'superiorparietal','superiortemporal','supramarginal','temporalpole',
 'transversetemporal'
]


# --- LH alignment ---
df_dk_mapped_lh = df_dk_mapped_lh.set_index('Region') if 'Region' in df_dk_mapped_lh.columns else df_dk_mapped_lh
zdiff_lh = zdiff_lh.set_index('Region')
zdiff_lh =  zdiff_lh.loc[DK_ORDER]
df_dk_mapped_lh = df_dk_mapped_lh.loc[DK_ORDER]    # reorder by DK order




zdiff_rh = zdiff_rh.set_index('Region')
# --- RH alignment ---
df_dk_mapped_rh = df_dk_mapped_rh.set_index('Region') if 'Region' in df_dk_mapped_rh.columns else df_dk_mapped_rh
zdiff_rh = zdiff_rh.loc[DK_ORDER]                # reorder by DK order
df_dk_mapped_rh = df_dk_mapped_rh.loc[DK_ORDER]



print("✅ Dataframes aligned to DK_ORDER successfully.")
print("LH shape:", zdiff_lh.shape, df_dk_mapped_lh.shape)
print("RH shape:", zdiff_rh.shape, df_dk_mapped_rh.shape)



# 3) LH centroids in the order of dk_parcalslh.index
coords_lh_list, regions_lh_kept = [], []
for region in df_dk_mapped_lh.index:
    lab = name_map_l.get(region)
    if lab is None: 
        continue
    mask = (dk_l == lab)
    if not mask.any():
        continue
    coords_lh_list.append(coords_lh[mask].mean(axis=0))
    regions_lh_kept.append(region)

# 4) RH centroids in the order of dk_parcalsrh.index
coords_rh_list, regions_rh_kept = [], []
for region in df_dk_mapped_rh.index:
    lab = name_map_r.get(region)
    if lab is None: 
        continue
    mask = (dk_r == lab)
    if not mask.any():
        continue
    coords_rh_list.append(coords_rh[mask].mean(axis=0))
    regions_rh_kept.append(region)




# 5) Align to regions available in your dataframes (PC1 loadings & ABAGEN means)
regions_lh_use = [r for r in regions_lh_kept if (r in zdiff_lh.index) and (r in df_dk_mapped_lh.index)]
regions_rh_use = [r for r in regions_rh_kept if (r in zdiff_rh.index) and (r in df_dk_mapped_rh.index)]



# slice centroid lists to match the "use" order
coords_lh_use = [c for r, c in zip(regions_lh_kept, coords_lh_list) if r in regions_lh_use]
coords_rh_use = [c for r, c in zip(regions_rh_kept, coords_rh_list) if r in regions_rh_use]

# 6) Build bilateral coords and vectors (same order!)
coords_bi = np.vstack([np.vstack(coords_lh_use), np.vstack(coords_rh_use)])






x_bi = np.concatenate([
    zdiff_lh.loc[regions_lh_use, 'PC1_Loading'].astype(float).values,
    zdiff_rh.loc[regions_rh_use, 'PC1_Loading'].astype(float).values
])

y_bi = np.concatenate([
    df_dk_mapped_lh.loc[regions_lh_use, 'PC1'].astype(float).values,
    df_dk_mapped_rh.loc[regions_rh_use, 'PC1'].astype(float).values
])

# basic checks
assert coords_bi.shape[0] == x_bi.shape[0] == y_bi.shape[0] and coords_bi.shape[0] > 0, "Alignment failed."

print(f"Using {len(regions_lh_use)} LH + {len(regions_rh_use)} RH = {len(x_bi)} parcels.")

# 7) BrainSMASH spatial null test (bilateral)
D = np.linalg.norm(coords_bi[:, None, :] - coords_bi[None, :, :], axis=-1)
gen = Base(x=y_bi, D=D, seed=42, resample=True)   # resample=True is usually more stable
n_perm = 10000
surrogates = gen(n=n_perm)

r_obs = np.corrcoef(x_bi, y_bi)[0, 1]
r_nulls = [np.corrcoef(x_bi, surrogates[i])[0, 1] for i in range(n_perm)]
pval = (np.sum(np.abs(r_nulls) >= np.abs(r_obs)) + 1) / (n_perm + 1)

print(f"[Bilateral] r = {r_obs:.3f} | BrainSMASH p = {pval:.5f} (n={n_perm})")



x_bi = np.concatenate([
    zdiff_lh.loc[regions_lh_use, 'PC1_Loading'].astype(float).values,
    zdiff_rh.loc[regions_rh_use, 'PC1_Loading'].astype(float).values
])

y_bi = np.concatenate([
    df_dk_mapped_lh.loc[regions_lh_use, 'PC2'].astype(float).values,
    df_dk_mapped_rh.loc[regions_rh_use, 'PC2'].astype(float).values
])

# basic checks
assert coords_bi.shape[0] == x_bi.shape[0] == y_bi.shape[0] and coords_bi.shape[0] > 0, "Alignment failed."

print(f"Using {len(regions_lh_use)} LH + {len(regions_rh_use)} RH = {len(x_bi)} parcels.")

# 7) BrainSMASH spatial null test (bilateral)
D = np.linalg.norm(coords_bi[:, None, :] - coords_bi[None, :, :], axis=-1)
gen = Base(x=y_bi, D=D, seed=42, resample=True)   # resample=True is usually more stable
n_perm = 10000
surrogates = gen(n=n_perm)

r_obs = np.corrcoef(x_bi, y_bi)[0, 1]
r_nulls = [np.corrcoef(x_bi, surrogates[i])[0, 1] for i in range(n_perm)]
pval = (np.sum(np.abs(r_nulls) >= np.abs(r_obs)) + 1) / (n_perm + 1)

print(f"[Bilateral] r = {r_obs:.3f} | BrainSMASH p = {pval:.5f} (n={n_perm})")



x_bi = np.concatenate([
    zdiff_lh.loc[regions_lh_use, 'PC1_Loading'].astype(float).values,
    zdiff_rh.loc[regions_rh_use, 'PC1_Loading'].astype(float).values
])

y_bi = np.concatenate([
    df_dk_mapped_lh.loc[regions_lh_use, 'PC3'].astype(float).values,
    df_dk_mapped_rh.loc[regions_rh_use, 'PC3'].astype(float).values
])

# basic checks
assert coords_bi.shape[0] == x_bi.shape[0] == y_bi.shape[0] and coords_bi.shape[0] > 0, "Alignment failed."

print(f"Using {len(regions_lh_use)} LH + {len(regions_rh_use)} RH = {len(x_bi)} parcels.")

# 7) BrainSMASH spatial null test (bilateral)
D = np.linalg.norm(coords_bi[:, None, :] - coords_bi[None, :, :], axis=-1)
gen = Base(x=y_bi, D=D, seed=42, resample=True)   # resample=True is usually more stable
n_perm = 10000
surrogates = gen(n=n_perm)

r_obs = np.corrcoef(x_bi, y_bi)[0, 1]
r_nulls = [np.corrcoef(x_bi, surrogates[i])[0, 1] for i in range(n_perm)]
pval = (np.sum(np.abs(r_nulls) >= np.abs(r_obs)) + 1) / (n_perm + 1)

print(f"[Bilateral] r = {r_obs:.3f} | BrainSMASH p = {pval:.5f} (n={n_perm})")




```
