```python
"""
Longitudinal PCA Analysis (PC1 Only)
------------------------------------
1. Loops through all Delta-Z files (V1-V2, V1-V3, ..., V3-V4).
2. Filters for specific Subcortical and Cortical regions.
3. Runs PCA to extract the FIRST Principal Component (PC1).
4. Saves:
   - PC1 Scores (The "Global Deviation Score" for each subject).
   - PC1 Loadings (Which brain regions drive this global score).
   - Variance Report.

Author: Niousha Dehestani
Date: 2026-02-12
"""

import pandas as pd
import numpy as np
import os
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# ===========================
# 1) CONFIGURATION
# ===========================
base_path = '/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/processt1/Results_Longitudinal'
output_dir = base_path # Saving in the same folder

# List of all Longitudinal Delta files to process
delta_files = [
    ('DeltaZ_V1_V2.csv', 'V1_V2'),
    ('DeltaZ_V1_V3.csv', 'V1_V3'),
    ('DeltaZ_V1_V4.csv', 'V1_V4'),
    ('DeltaZ_V2_V3.csv', 'V2_V3'),
    ('DeltaZ_V2_V4.csv', 'V2_V4'),
    ('DeltaZ_V3_V4.csv', 'V3_V4')
]

# ROI Dictionary (Features to include)
vol_categories = {
    'mr_y_smri__vol__aseg__ag__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__ag__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__cd__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__cd__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__hc__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__hc__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__pl__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__pl__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__pt__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__pt__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__th__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__th__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__ab__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__ab__rh_sum':'Subcortical_Right',
    
    'mr_y_smri__vol__dsk__ins__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__stmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__ttmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__bstmps__lh_sum':'Temporal_Left',    
    'mr_y_smri__vol__dsk__ptmp__lh_sum':'Temporal_Left',  
    'mr_y_smri__vol__dsk__er__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__mtmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__ff__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__itmp__lh_sum':'Temporal_Left', 
    'mr_y_smri__vol__dsk__ph__lh_sum':'Temporal_Left',
    
    'mr_y_smri__vol__dsk__pfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__prcn__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pactr__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__mobfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__lobfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pop__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pob__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__ptg__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__cmfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__cac__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__prctr__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__sfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__rmfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__rac__lh_sum':'Frontal_Left',
    
    'mr_y_smri__vol__dsk__iprt__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__pcg__lh_sum':'Parietal_Left',  
    'mr_y_smri__vol__dsk__poctr__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__ic__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__pcc__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__locc__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__lg__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__cn__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__sm__lh_sum':'Parietal_Left',    
    'mr_y_smri__vol__dsk__sprt__lh_sum':'Parietal_Left',
    
    'mr_y_smri__vol__dsk__bstmps__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ptmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ttmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__er__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__mtmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ff__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__itmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__stmp__rh_sum':'Temporal_Right',    
    'mr_y_smri__vol__dsk__ph__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ins__rh_sum':'Temporal_Right',
    
    'mr_y_smri__vol__dsk__pfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__prctr__rh_sum':'Frontal_Right', 
    'mr_y_smri__vol__dsk__pactr__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__mobfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__lobfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__pop__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__pob__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__ptg__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__rac__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__cmfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__sfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__rmfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__cac__rh_sum':'Frontal_Right',    
    
    'mr_y_smri__vol__dsk__sprt__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__iprt__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__poctr__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__pcg__rh_sum':'Parietal_Right',  
    'mr_y_smri__vol__dsk__prcn__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__sm__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__ic__rh_sum':'Parietal_Right',
    
    'mr_y_smri__vol__dsk__pcc__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__locc__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__lg__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__cn__rh_sum':'Occipital_Right'
}

# ===========================
# 2) PIPELINE FUNCTION
# ===========================
def run_pc1_analysis(input_path, output_dir, label_suffix):
    print(f"\n🚀 Processing: {label_suffix}")
    
    # Check if file exists
    if not os.path.exists(input_path):
        print(f"   ⚠️ File not found: {input_path}")
        return

    # Load Data (Index is 'ID' from the previous fix)
    df = pd.read_csv(input_path).set_index('ID')
    
    # Keep 'group' for later, but remove from matrix
    subject_groups = df['group'].copy() if 'group' in df.columns else None
    
    # Filter for ROI columns only
    valid_cols = [c for c in df.columns if c in vol_categories.keys()]
    df_numeric = df[valid_cols].dropna()
    
    if df_numeric.empty:
        print("   ❌ Error: Dataframe is empty after filtering/dropping NaNs.")
        return
        
    print(f"   Subjects: {len(df_numeric)} | ROIs: {len(valid_cols)}")

    # Standardize (Z-score the Delta-Zs)
    scaler = StandardScaler()
    df_scaled = scaler.fit_transform(df_numeric)

    # PCA (PC1 Only)
    pca = PCA(n_components=1)
    pca_scores = pca.fit_transform(df_scaled)
    
    # Variance Check
    variance_explained = pca.explained_variance_ratio_[0]
    print(f"   📊 PC1 Explained Variance: {variance_explained:.2%}")

    # 1. Save Scores (Subject Level)
    df_scores = pd.DataFrame(
        pca_scores, 
        columns=['PC1'], 
        index=df_numeric.index
    )
    # Add group column back
    if subject_groups is not None:
        df_scores = df_scores.join(subject_groups, how='left')
        
    score_file = os.path.join(output_dir, f"PC1_Scores_{label_suffix}.csv")
    df_scores.to_csv(score_file)
    print(f"   ✅ Saved Scores: {score_file}")

    # 2. Save Loadings (ROI Level)
    df_loadings = pd.DataFrame(
        pca.components_.T, 
        index=valid_cols, 
        columns=['PC1_Loading']
    )
    # Add region label (Lobe/Subcortical)
    df_loadings['Region'] = df_loadings.index.map(vol_categories)
    
    loading_file = os.path.join(output_dir, f"PC1_Loadings_{label_suffix}.csv")
    df_loadings.sort_values('PC1_Loading', ascending=False).to_csv(loading_file)
    print(f"   ✅ Saved Loadings: {loading_file}")

# ===========================
# 3) EXECUTION LOOP
# ===========================
print("--- Starting Longitudinal PCA (PC1 Only) ---")

for fname, suffix in delta_files:
    full_path = os.path.join(base_path, fname)
    run_pc1_analysis(full_path, output_dir, suffix)

print("\n✨ All files processed.")

```

```python
"""
Cross-Sectional PCA Analysis (V1, V2, V3)
-----------------------------------------
1. Loads the master 'Z_all_timepoints.csv' file.
2. Splits data into V1, V2, V3 based on session IDs.
3. Runs PCA (PC1 Only) for each visit independently.
4. Saves PC1 Scores (Subject-level) and Loadings (ROI-level).

Author: Niousha Dehestani
Date: 2026-02-12
"""

import pandas as pd
import numpy as np
import os
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# ===========================
# 1) CONFIGURATION
# ===========================
base_dir = "/Users/nioushad/Documents/myJupyter/normative modeling/code/component//release6/hpc/finalpaper/output/processt1/Results_Longitudinal"
z_score_file = os.path.join(base_dir, "Z_all_timepoints.csv")

# Output Directory
output_dir = os.path.join(base_dir, "PCA_CrossSectional")
os.makedirs(output_dir, exist_ok=True)

# ROI Dictionary (Features to include)
vol_categories = {
    'mr_y_smri__vol__aseg__ag__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__ag__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__cd__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__cd__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__hc__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__hc__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__pl__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__pl__rh_sum':'Subcortical_Right',
    'mr_y_smri__vol__aseg__pt__lh_sum':'Subcortical_Left',
    'mr_y_smri__vol__aseg__pt__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__th__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__th__rh_sum': 'Subcortical_Right',
    'mr_y_smri__vol__aseg__ab__lh_sum': 'Subcortical_Left',
    'mr_y_smri__vol__aseg__ab__rh_sum':'Subcortical_Right',
    
    'mr_y_smri__vol__dsk__ins__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__stmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__ttmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__bstmps__lh_sum':'Temporal_Left',    
    'mr_y_smri__vol__dsk__ptmp__lh_sum':'Temporal_Left',  
    'mr_y_smri__vol__dsk__er__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__mtmp__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__ff__lh_sum':'Temporal_Left',
    'mr_y_smri__vol__dsk__itmp__lh_sum':'Temporal_Left', 
    'mr_y_smri__vol__dsk__ph__lh_sum':'Temporal_Left',
    
    'mr_y_smri__vol__dsk__pfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__prcn__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pactr__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__mobfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__lobfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pop__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__pob__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__ptg__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__cmfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__cac__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__prctr__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__sfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__rmfrt__lh_sum':'Frontal_Left',
    'mr_y_smri__vol__dsk__rac__lh_sum':'Frontal_Left',
    
    'mr_y_smri__vol__dsk__iprt__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__pcg__lh_sum':'Parietal_Left',  
    'mr_y_smri__vol__dsk__poctr__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__ic__lh_sum':'Parietal_Left',
    'mr_y_smri__vol__dsk__pcc__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__locc__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__lg__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__cn__lh_sum':'Occipital_Left',
    'mr_y_smri__vol__dsk__sm__lh_sum':'Parietal_Left',    
    'mr_y_smri__vol__dsk__sprt__lh_sum':'Parietal_Left',
    
    'mr_y_smri__vol__dsk__bstmps__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ptmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ttmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__er__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__mtmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ff__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__itmp__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__stmp__rh_sum':'Temporal_Right',    
    'mr_y_smri__vol__dsk__ph__rh_sum':'Temporal_Right',
    'mr_y_smri__vol__dsk__ins__rh_sum':'Temporal_Right',
    
    'mr_y_smri__vol__dsk__pfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__prctr__rh_sum':'Frontal_Right', 
    'mr_y_smri__vol__dsk__pactr__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__mobfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__lobfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__pop__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__pob__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__ptg__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__rac__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__cmfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__sfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__rmfrt__rh_sum':'Frontal_Right',
    'mr_y_smri__vol__dsk__cac__rh_sum':'Frontal_Right',    
    
    'mr_y_smri__vol__dsk__sprt__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__iprt__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__poctr__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__pcg__rh_sum':'Parietal_Right',  
    'mr_y_smri__vol__dsk__prcn__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__sm__rh_sum':'Parietal_Right',
    'mr_y_smri__vol__dsk__ic__rh_sum':'Parietal_Right',
    
    'mr_y_smri__vol__dsk__pcc__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__locc__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__lg__rh_sum':'Occipital_Right',
    'mr_y_smri__vol__dsk__cn__rh_sum':'Occipital_Right'
}

# ===========================
# 2) LOAD DATA
# ===========================
print("📊 Loading Z-score Master File...")
if not os.path.exists(z_score_file):
    print(f"❌ Error: Could not find {z_score_file}")
    exit()

df_z = pd.read_csv(z_score_file)
# Ensure ID-wave is the index
if 'ID-wave' in df_z.columns:
    df_z = df_z.set_index('ID-wave')
elif 'Unnamed: 0' in df_z.columns:
    df_z = df_z.rename(columns={'Unnamed: 0': 'ID-wave'}).set_index('ID-wave')

print(f"   Loaded: {len(df_z)} total records.")

# ===========================
# 3) PCA FUNCTION
# ===========================
def run_visit_pca(df_subset, visit_label):
    print(f"\n🚀 Processing: {visit_label}")
    
    # Filter for ROI columns
    valid_cols = [c for c in df_subset.columns if c in vol_categories.keys()]
    
    # 1. Clean Data (Numeric only, no NaNs)
    df_numeric = df_subset[valid_cols].dropna()
    
    if df_numeric.empty:
        print(f"   ⚠️ No data found for {visit_label}. Skipping.")
        return

    print(f"   Subjects: {len(df_numeric)} | ROIs: {len(valid_cols)}")

    # 2. Standardize
    # 
    scaler = StandardScaler()
    df_scaled = scaler.fit_transform(df_numeric)

    # 3. PCA (PC1 Only)
    pca = PCA(n_components=1)
    pca_scores = pca.fit_transform(df_scaled)
    
    var_explained = pca.explained_variance_ratio_[0]
    print(f"   📊 PC1 Explained Variance: {var_explained:.2%}")

    # 4. Save Scores (Subject Level)
    # We clean the index to be just 'participant_id' (remove _ses-00A)
    clean_index = df_numeric.index.str.split('_ses').str[0]
    
    df_scores = pd.DataFrame(
        pca_scores, 
        columns=['PC1'], 
        index=clean_index
    )
    df_scores.index.name = 'ID'
    
    score_file = os.path.join(output_dir, f"PC1_Scores_{visit_label}.csv")
    df_scores.to_csv(score_file)
    print(f"   ✅ Saved Scores: {score_file}")

    # 5. Save Loadings (ROI Level)
    df_loadings = pd.DataFrame(
        pca.components_.T, 
        index=valid_cols, 
        columns=['PC1_Loading']
    )
    df_loadings['Region'] = df_loadings.index.map(vol_categories)
    
    loading_file = os.path.join(output_dir, f"PC1_Loadings_{visit_label}.csv")
    df_loadings.sort_values('PC1_Loading', ascending=False).to_csv(loading_file)
    print(f"   ✅ Saved Loadings: {loading_file}")

# ===========================
# 4) EXECUTION LOOP
# ===========================
visits = {
    'V1': 'ses-00A',
    'V2': 'ses-02A',
    'V3': 'ses-04A',
    'V4': 'ses-06A' # Optional, included just in case
}

for v_label, session_id in visits.items():
    # Filter dataframe for this specific session
    df_visit = df_z[df_z.index.str.contains(session_id)].copy()
    
    # Run PCA
    run_visit_pca(df_visit, v_label)

print("\n✨ All Cross-Sectional PCA runs finished.")
```
