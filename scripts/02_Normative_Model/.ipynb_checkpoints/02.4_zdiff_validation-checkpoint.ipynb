# 07 — Validate z-diff Against Observed Longitudinal Change

Addresses the concern that a cross-sectional normative model may not faithfully capture individual longitudinal change (Reviewer Comment M1).

For each held-out healthy subject with paired timepoints we compute:
- **Expected change**: normative model predicted value at T2 − predicted value at T1 (from `Yhat`)
- **Observed change**: actual brain volume at T2 − T1 (from `Y`)

Then per region we report:
- Pearson r between expected and observed change
- Mean absolute error (MAE)

**Input:** `outputs/longitudinal_healthy/` — `Y_all_timepoints_healthy.csv` and `Yhat_all_timepoints_healthy.csv` (output of `06_longitudinal_predict_healthy.ipynb`)  
**Outputs:** `zdiff_validation_summary.csv`, `zdiff_validation_per_ROI.csv`, scatter plots for representative regions

```python
# ── CONFIG ──────────────────────────────────────────────────────────────────
RESULTS_DIR = 'outputs/longitudinal_healthy'
OUTPUT_DIR  = 'outputs/zdiff_validation'

# Timepoint pairs to evaluate  (session_id_start, session_id_end, label)
VISIT_PAIRS = [
    ('ses-00A', 'ses-02A', 'V1_V2'),
    ('ses-02A', 'ses-04A', 'V2_V3'),
    ('ses-00A', 'ses-04A', 'V1_V3'),
]

# Number of representative regions to plot (ranked by median |r| across pairs)
N_EXAMPLE_REGIONS = 6
```

```python
import os
import numpy as np
import pandas as pd
from scipy.stats import pearsonr
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import seaborn as sns

os.makedirs(OUTPUT_DIR, exist_ok=True)
```

## Step 1 — Load predicted and observed values

```python
df_Y    = pd.read_csv(os.path.join(RESULTS_DIR, 'Y_all_timepoints_healthy.csv'),    index_col=0)
df_Yhat = pd.read_csv(os.path.join(RESULTS_DIR, 'Yhat_all_timepoints_healthy.csv'), index_col=0)

roi_ids = df_Y.columns.tolist()
print(f'Loaded: {df_Y.shape[0]} rows | {len(roi_ids)} ROIs')
print(f'Index sample: {df_Y.index[:3].tolist()}')
```

## Step 2 — Compute expected and observed change per visit pair

```python
def extract_visit(df, visit_id):
    """Return subject-indexed subset for one visit, stripping the session suffix."""
    sub = df[df.index.str.contains(visit_id)].copy()
    sub.index = sub.index.str.split('_ses').str[0]
    return sub[~sub.index.duplicated(keep='first')]


pair_data = {}
for vis_start, vis_end, label in VISIT_PAIRS:
    Y1    = extract_visit(df_Y,    vis_start)
    Y2    = extract_visit(df_Y,    vis_end)
    Yhat1 = extract_visit(df_Yhat, vis_start)
    Yhat2 = extract_visit(df_Yhat, vis_end)

    common = Y1.index.intersection(Y2.index).intersection(Yhat1.index).intersection(Yhat2.index)
    if len(common) < 5:
        print(f'  Skipping {label}: only {len(common)} subjects with complete data')
        continue

    obs_change  = (Y2.loc[common]    - Y1.loc[common]).astype(float)
    exp_change  = (Yhat2.loc[common] - Yhat1.loc[common]).astype(float)
    pair_data[label] = {'obs': obs_change, 'exp': exp_change, 'n': len(common)}
    print(f'  {label}: {len(common)} subjects')
```

## Step 3 — Per-region Pearson r and MAE

```python
records = []
for label, d in pair_data.items():
    obs = d['obs']
    exp = d['exp']
    for roi in roi_ids:
        mask = obs[roi].notna() & exp[roi].notna()
        if mask.sum() < 5:
            continue
        o = obs.loc[mask, roi].values
        e = exp.loc[mask, roi].values
        r, p = pearsonr(o, e)
        mae  = np.mean(np.abs(o - e))
        records.append({'pair': label, 'ROI': roi, 'r': r, 'p': p,
                        'MAE': mae, 'n': int(mask.sum())})

df_roi = pd.DataFrame(records)
df_roi.to_csv(os.path.join(OUTPUT_DIR, 'zdiff_validation_per_ROI.csv'), index=False)

summary = (
    df_roi.groupby('pair')[['r', 'MAE']]
    .agg(['mean', 'median', 'std'])
    .round(4)
)
summary.columns = ['_'.join(c) for c in summary.columns]
summary.to_csv(os.path.join(OUTPUT_DIR, 'zdiff_validation_summary.csv'))

print(summary.to_string())
```

## Step 4 — Distribution of per-region r across visit pairs

```python
pairs = sorted(df_roi['pair'].unique())
fig, axes = plt.subplots(1, len(pairs), figsize=(5 * len(pairs), 4), sharey=False)
if len(pairs) == 1:
    axes = [axes]

for ax, label in zip(axes, pairs):
    r_vals = df_roi[df_roi['pair'] == label]['r']
    ax.hist(r_vals, bins=20, color='steelblue', edgecolor='white', alpha=0.85)
    ax.axvline(r_vals.median(), color='darkorange', linewidth=2,
               label=f'median r = {r_vals.median():.2f}')
    ax.axvline(0, color='grey', linewidth=1, linestyle='--')
    ax.set_xlabel('Pearson r (expected vs observed change)', fontsize=11)
    ax.set_ylabel('Number of ROIs', fontsize=11)
    ax.set_title(label, fontsize=12, fontweight='bold')
    ax.legend(fontsize=10)
    sns.despine(ax=ax)

plt.suptitle('Expected vs Observed Brain Volume Change — Per-ROI Correlation', fontsize=13)
plt.tight_layout()
fig.savefig(os.path.join(OUTPUT_DIR, 'zdiff_validation_r_distribution.png'), dpi=300, bbox_inches='tight')
plt.close(fig)
print('Saved: zdiff_validation_r_distribution.png')
```

## Step 5 — Scatter plots for representative regions

Regions are ranked by median |r| across all visit pairs; the top `N_EXAMPLE_REGIONS` are plotted.

```python
# Pick top regions by median absolute r across all pairs
top_rois = (
    df_roi.groupby('ROI')['r']
    .apply(lambda x: x.abs().median())
    .nlargest(N_EXAMPLE_REGIONS)
    .index.tolist()
)

for label, d in pair_data.items():
    obs = d['obs']
    exp = d['exp']

    ncols = 3
    nrows = int(np.ceil(N_EXAMPLE_REGIONS / ncols))
    fig, axes = plt.subplots(nrows, ncols, figsize=(5 * ncols, 4.5 * nrows))
    axes = axes.flatten()

    for ax, roi in zip(axes, top_rois):
        mask = obs[roi].notna() & exp[roi].notna()
        o = obs.loc[mask, roi].values
        e = exp.loc[mask, roi].values
        r_val = df_roi[(df_roi['pair'] == label) & (df_roi['ROI'] == roi)]['r'].values
        r_txt = f'r = {r_val[0]:.2f}' if len(r_val) else ''

        sns.regplot(x=e, y=o, ax=ax,
                    scatter_kws={'alpha': 0.4, 'color': '#444444', 's': 30},
                    line_kws={'color': 'darkorange', 'linewidth': 2})
        ax.set_xlabel('Expected change (Δŷ)', fontsize=10)
        ax.set_ylabel('Observed change (Δy)', fontsize=10)
        ax.set_title(roi.replace('mr_y_smri__vol__dsk__', '').replace('_sum', ''),
                     fontsize=9, fontweight='bold')
        ax.annotate(r_txt, xy=(0.05, 0.88), xycoords='axes fraction', fontsize=10)
        sns.despine(ax=ax)

    for ax in axes[N_EXAMPLE_REGIONS:]:
        ax.set_visible(False)

    plt.suptitle(f'Expected vs Observed Change — {label}', fontsize=13, fontweight='bold')
    plt.tight_layout()
    fig.savefig(os.path.join(OUTPUT_DIR, f'zdiff_validation_scatter_{label}.png'),
                dpi=300, bbox_inches='tight')
    plt.close(fig)
    print(f'Saved: zdiff_validation_scatter_{label}.png')

print('\nValidation complete.')
print(f'Per-ROI results: {os.path.join(OUTPUT_DIR, "zdiff_validation_per_ROI.csv")}')
print(f'Summary:         {os.path.join(OUTPUT_DIR, "zdiff_validation_summary.csv")}')
```
