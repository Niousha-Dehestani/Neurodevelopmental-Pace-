```python
"""
Cross-Sectional Normative Model (BLR)
------------------------------------
- Implementation: Bayesian Linear Regression with B-Spline Basis
- Pipeline: Feature Splitting -> Basis Expansion -> Estimation -> Site Evaluation
- Inputs: Final Training Split (Family Independent)
- Outputs: Z-scores, Global Metrics, and Site-Specific Performance
- Filter: Site Evaluation only for N >= 10 (Test Set)

Author: Niousha Dehestani
Date: 2026-02-10
"""

import pandas as pd
import numpy as np
import os
import warnings
from scipy.linalg import LinAlgWarning
from sklearn.model_selection import train_test_split
from pcntoolkit.normative import estimate, evaluate
from pcntoolkit.util.utils import create_bspline_basis

# ===========================
# 1) CONFIGURATION & LOADING
# ===========================
input_file = '/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/final_train2.csv'
output_dir = '/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/processt1'
os.makedirs(output_dir, exist_ok=True)
os.makedirs(os.path.join(output_dir, 'ROI_models'), exist_ok=True)

# Load Data
all_data = pd.read_csv(input_file).set_index('ID-wave')
all_data = all_data.drop(columns =[
 'mr_y_smri__vol__dsk__lh_sum',
 'mr_y_smri__vol__dsk__rh_sum',
 'mr_y_smri__vol__dsk_sum',
])


# Site Mapping
site_mapping = {
    'HASH4d1ed7b1_9':1, 'HASH3935c89e_16':2, 'HASHd7cb4c6d_10':3, 'HASHe3ce02d3_10':4,
    'HASH7911780b_15':5, 'HASH3ce956bd_18':6, 'HASHa3e45734_18':7, 'HASH4036a433_21':8,
    'HASHdb2589d4_17':9, 'HASH1314a204_2':10, 'HASHfeb7e81a_4':11, 'HASHc3bf3d9c_13':12,
    'HASH69f406fa_13':13, 'HASH96a0c182_6':14, 'HASHe4f6957a_12':15, 'HASH4b0b8b05_4':16,
    'HASH11ad4ed5_14':17, 'HASH7f91147d_14':18, 'HASH65b39280_7':19, 'HASH311170b9_5':20,
    'HASHe76e6d72_21':21, 'HASHb640a1b8_21':22, 'HASH03db707f_11':23, 'HASHd422be27_20':24,
    'HASHc9398971_20':25, 'HASH5b0cf1bb_3':26, 'HASH5b2fcf80_8':27, 'HASH5ac2b20b_19':28,
    'HASHc1b3365b_13':29, 'HASH6b4422a7_1':30
}
all_data['site_mri'] = all_data['site_mri'].map(site_mapping).astype('Int64')

# ===========================
# 2) FEATURE & COVARIATE PREP
# ===========================
roi_ids = all_data.columns[all_data.columns.str.contains('mr_y_smri__')].tolist()

all_data_features = all_data[roi_ids]
all_data_covariates = all_data[['ab_g_dyn__visit_age', 'ab_g_stc__cohort_sex', 'site_mri']]

# One-hot encoding sites
all_data_covariates = pd.get_dummies(all_data_covariates, columns=['site_mri'], dtype=int)

# ===========================
# 3) TRAIN/TEST SPLIT
# ===========================
X_train, X_test, y_train, y_test = train_test_split(
    all_data_covariates,
    all_data_features,
    test_size=0.2,
    random_state=42,
)

# 4) Reset indexes for safe slicing
for df in [X_train, X_test, y_train, y_test]:
    df.reset_index(drop=True, inplace=True)

# 5) Build integer-based site splits
sites = [
    X_test.index[X_test[f'site_mri_{i}'] == 1].tolist()
    for i in range(1, 31)
]
site_names = [f"site_{i}" for i in range(1, 31)]

# ===========================
# 6) SAVE BASE FILES (INTERCEPT ONLY)
# ===========================
y_train.to_csv(os.path.join(output_dir, 'resp_tr.txt'), sep='\t', header=False, index=False)
y_test.to_csv(os.path.join(output_dir, 'resp_te.txt'), sep='\t', header=False, index=False)
X_train.to_csv(os.path.join(output_dir, 'cov_tr.txt'), sep='\t', header=False, index=False)
X_test.to_csv(os.path.join(output_dir, 'cov_te.txt'), sep='\t', header=False, index=False)

for c in y_train.columns:
    y_train[c].to_csv(os.path.join(output_dir, f'resp_tr_{c}.txt'), header=False, index=False)
    y_test[c].to_csv(os.path.join(output_dir, f'resp_te_{c}.txt'), header=False, index=False)

# ===========================
# 7) DEFINE B-SPLINE BASIS
# ===========================
xmin, xmax = 7, 18
B = create_bspline_basis(xmin, xmax, nknots=3)

# ===========================
# 8) INITIALIZE METRICS DATAFRAMES
# ===========================
blr_metrics = pd.DataFrame(columns=['ROI','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
blr_site_metrics = pd.DataFrame(columns=['ROI','site','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
blr_globmetrics = pd.DataFrame(columns=['ROI','site','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
metrics_globmetrics = pd.DataFrame(columns=['ROI','yhat_te','s2_te','Z'])

# ===========================
# 9) ROI MODELING LOOP
# ===========================
# Suppress specific warnings for clean output
warnings.filterwarnings("ignore", category=LinAlgWarning)
warnings.filterwarnings("ignore", category=RuntimeWarning)

for roi in roi_ids:
    print(f'🚀 Running ROI: {roi}')
    roi_dir = os.path.join(output_dir, roi)
    os.makedirs(roi_dir, exist_ok=True)
    os.makedirs(os.path.join(roi_dir, 'blr'), exist_ok=True)
    
    # 9a) Load base covariates for expansion
    X_tr = np.loadtxt(os.path.join(output_dir, 'cov_tr.txt'))
    X_te = np.loadtxt(os.path.join(output_dir, 'cov_te.txt'))

    # 9b) Add Intercept
    X_tr = np.hstack([X_tr, np.ones((X_tr.shape[0], 1))])
    X_te = np.hstack([X_te, np.ones((X_te.shape[0], 1))])

    # Save intercept-only versions
    np.savetxt(os.path.join(output_dir, 'cov_int_tr.txt'), X_tr)
    np.savetxt(os.path.join(output_dir, 'cov_int_te.txt'), X_te)

    # 9c) Construct Spline Basis
    Phi_tr = np.vstack([B(age) for age in X_tr[:, 0]])
    Phi_te = np.vstack([B(age) for age in X_te[:, 0]])

    # Append to design matrix
    X_tr = np.hstack([X_tr, Phi_tr])
    X_te = np.hstack([X_te, Phi_te])

    # Save expanded covariates
    cov_tr_path = os.path.join(roi_dir, 'cov_bspline_tr.txt')
    cov_te_path = os.path.join(roi_dir, 'cov_bspline_te.txt')
    np.savetxt(cov_tr_path, X_tr)
    np.savetxt(cov_te_path, X_te)

    # 9d) Prepare Response Paths
    resp_tr_path = os.path.join(output_dir, f'resp_tr_{roi}.txt')
    resp_te_path = os.path.join(output_dir, f'resp_te_{roi}.txt')

    # Change directory for PCNToolkit safety
    os.chdir(roi_dir)

    try:
        # 9e) ESTIMATE MODEL
        yhat_te, s2_te, nm, Z, metrics_te = estimate(
            cov_tr_path, resp_tr_path,
            testresp=resp_te_path, testcov=cov_te_path,
            alg='blr', optimizer='powell',
            cv_folds=5,
            savemodel=True, saveoutput=False,
            standardize=False,
            prop_main=[i for i in range(X_tr.shape[1])],
            all_covariates=[i for i in range(X_tr.shape[1])]
        )

        # Save outputs
        np.savetxt(f'{roi}_yhat_te.txt', yhat_te)
        np.savetxt(f'{roi}_s2_te.txt', s2_te)
        np.savetxt(f'{roi}_Z.txt', Z)
        print(f"Saved outputs for {roi}")

        # Collect Prediction Data
        metrics_globmetrics.loc[len(metrics_globmetrics)] = [
            roi,
            yhat_te.flatten().tolist(),
            s2_te.flatten().tolist(),
            Z.flatten().tolist()
        ]

        # Collect Test Metrics
        blr_metrics.loc[len(blr_metrics)] = [
            roi,
            metrics_te['MSLL'][0], metrics_te['EXPV'][0],
            metrics_te['SMSE'][0], metrics_te['RMSE'][0],
            metrics_te['Rho'][0],  metrics_te['pRho'][0],
        ]

        # 9f) GLOBAL EVALUATION
        y_te = np.loadtxt(resp_te_path)[:, None]
        y_tr = np.loadtxt(resp_tr_path)[:, None]

        metricsglob = evaluate(
            y_te, yhat_te, s2_te,
            np.array([y_tr.mean()]), np.array([y_tr.std()])
        )

        blr_globmetrics.loc[len(blr_globmetrics)] = [
            roi, 'global',
            metricsglob['MSLL'][0], metricsglob['EXPV'][0],
            metricsglob['SMSE'][0], metricsglob['RMSE'][0],
            metricsglob['Rho'][0],  metricsglob['pRho'][0],
        ]

        # 9g) SITE-LEVEL EVALUATION (Filtered N >= 10)
        # 
        for num, site_idx in enumerate(sites):
            # ------------------------------------------------------------------
            # STATISTICAL FILTER: Only evaluate sites with >= 10 subjects in Test
            # ------------------------------------------------------------------
            if len(site_idx) < 10: 
                continue
            
            try:
                mets = evaluate(
                    y_te[site_idx], yhat_te[site_idx], s2_te[site_idx],
                    np.array([y_tr.mean()]), np.array([y_tr.std()])
                )
                
                blr_site_metrics.loc[len(blr_site_metrics)] = [
                    roi, site_names[num],
                    mets['MSLL'][0], mets['EXPV'][0],
                    mets['SMSE'][0], mets['RMSE'][0],
                    mets['Rho'][0],  mets['pRho'][0],
                ]
            except Exception:
                continue

    except Exception as e:
        print(f"   ❌ Error processing {roi}: {str(e)}")

# ===========================
# 10) SAVE FINAL RESULTS
# ===========================
os.chdir(output_dir) # Return to main output dir
blr_metrics.to_csv('/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/blr_metrics_new.csv', index=False)
blr_site_metrics.to_csv('/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/blr_site_metrics.csv', index=False)
blr_globmetrics.to_csv('/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/blr_globmetrics_new.csv', index=False)
metrics_globmetrics.to_csv('/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/hpc/finalpaper/output/blr_yhat_s2_Z.csv', index=False)

print("✅ Pipeline Finished. Site metrics saved for N >= 10.")
```

```python
"""
Cross-Sectional Normative Model (BLR)
------------------------------------
- Implementation: Bayesian Linear Regression with B-Spline Basis
- Pipeline: Feature Splitting -> Basis Expansion -> Estimation -> Site Evaluation
- Inputs: Final Training Split (Family Independent)
- Outputs: Z-scores, Global Metrics, and Site-Specific Performance

Author: Niousha Dehestani
Date: 2026-02-10
"""

import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split
from pcntoolkit.normative import estimate, evaluate
from pcntoolkit.util.utils import create_bspline_basis

# ===========================
# 1) CONFIGURATION & LOADING
# ===========================
input_file = '/Users/nioushad/Documents/normative modeling/data/splits/final_train2.csv'
output_dir = '/Users/nioushad/Documents/myJupyter/normative modeling/code/component/release6/processt1'
os.makedirs(output_dir, exist_ok=True)
os.makedirs(os.path.join(output_dir, 'ROI_models'), exist_ok=True)

# Load Data
all_data = pd.read_csv(input_file).set_index('ID-wave')

# Site Mapping
site_mapping = {
    'HASH4d1ed7b1_9':1, 'HASH3935c89e_16':2, 'HASHd7cb4c6d_10':3, 'HASHe3ce02d3_10':4,
    'HASH7911780b_15':5, 'HASH3ce956bd_18':6, 'HASHa3e45734_18':7, 'HASH4036a433_21':8,
    'HASHdb2589d4_17':9, 'HASH1314a204_2':10, 'HASHfeb7e81a_4':11, 'HASHc3bf3d9c_13':12,
    'HASH69f406fa_13':13, 'HASH96a0c182_6':14, 'HASHe4f6957a_12':15, 'HASH4b0b8b05_4':16,
    'HASH11ad4ed5_14':17, 'HASH7f91147d_14':18, 'HASH65b39280_7':19, 'HASH311170b9_5':20,
    'HASHe76e6d72_21':21, 'HASHb640a1b8_21':22, 'HASH03db707f_11':23, 'HASHd422be27_20':24,
    'HASHc9398971_20':25, 'HASH5b0cf1bb_3':26, 'HASH5b2fcf80_8':27, 'HASH5ac2b20b_19':28,
    'HASHc1b3365b_13':29, 'HASH6b4422a7_1':30
}
all_data['site_mri'] = all_data['site_mri'].map(site_mapping).astype('Int64')

# ===========================
# 2) FEATURE & COVARIATE PREP
# ===========================
roi_ids = all_data.columns[all_data.columns.str.contains('mr_y_smri__')].tolist()

all_data_features = all_data[roi_ids]
all_data_covariates = all_data[['ab_g_dyn__visit_age', 'ab_g_stc__cohort_sex', 'site_mri']]

# One-hot encoding sites
all_data_covariates = pd.get_dummies(all_data_covariates, columns=['site_mri'], dtype=int)

# ===========================
# 3) TRAIN/TEST SPLIT
# ===========================
X_train, X_test, y_train, y_test = train_test_split(
    all_data_covariates,
    all_data_features,
    test_size=0.2,
    random_state=42,
)

# 4) Reset indexes for safe slicing
for df in [X_train, X_test, y_train, y_test]:
    df.reset_index(drop=True, inplace=True)

# 5) Build integer-based site splits
sites = [
    X_test.index[X_test[f'site_mri_{i}'] == 1].tolist()
    for i in range(1, 31)
]
site_names = [f"site_{i}" for i in range(1, 31)]

# ===========================
# 6) SAVE BASE FILES (INTERCEPT ONLY)
# ===========================
# Save overall train/test response & covariates
y_train.to_csv(os.path.join(output_dir, 'resp_tr.txt'), sep='\t', header=False, index=False)
y_test.to_csv(os.path.join(output_dir, 'resp_te.txt'), sep='\t', header=False, index=False)
X_train.to_csv(os.path.join(output_dir, 'cov_tr.txt'), sep='\t', header=False, index=False)
X_test.to_csv(os.path.join(output_dir, 'cov_te.txt'), sep='\t', header=False, index=False)

# Save per-ROI response files
for c in y_train.columns:
    y_train[c].to_csv(os.path.join(output_dir, f'resp_tr_{c}.txt'), header=False, index=False)
    y_test[c].to_csv(os.path.join(output_dir, f'resp_te_{c}.txt'), header=False, index=False)

# ===========================
# 7) DEFINE B-SPLINE BASIS
# ===========================
xmin, xmax = 7, 18
B = create_bspline_basis(xmin, xmax, nknots=3)

# ===========================
# 8) INITIALIZE METRICS DATAFRAMES
# ===========================
blr_metrics = pd.DataFrame(columns=['ROI','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
blr_site_metrics = pd.DataFrame(columns=['ROI','site','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
blr_globmetrics = pd.DataFrame(columns=['ROI','site','MSLL','EXPV','SMSE','RMSE','Rho','pRho'])
metrics_globmetrics = pd.DataFrame(columns=['ROI','yhat_te','s2_te','Z'])

# ===========================
# 9) ROI MODELING LOOP
# ===========================
for roi in roi_ids:
    print(f'🚀 Running ROI: {roi}')
    roi_dir = os.path.join(output_dir, roi)
    os.makedirs(roi_dir, exist_ok=True)
    os.makedirs(os.path.join(roi_dir, 'blr'), exist_ok=True)
    
    # 9a) Load base covariates for expansion
    X_tr = np.loadtxt(os.path.join(output_dir, 'cov_tr.txt'))
    X_te = np.loadtxt(os.path.join(output_dir, 'cov_te.txt'))

    # 9b) Add Intercept
    X_tr = np.hstack([X_tr, np.ones((X_tr.shape[0], 1))])
    X_te = np.hstack([X_te, np.ones((X_te.shape[0], 1))])

    # Save intercept-only versions
    np.savetxt(os.path.join(output_dir, 'cov_int_tr.txt'), X_tr)
    np.savetxt(os.path.join(output_dir, 'cov_int_te.txt'), X_te)

    # 9c) Construct Spline Basis (Age is column 0)
    Phi_tr = np.vstack([B(age) for age in X_tr[:, 0]])
    Phi_te = np.vstack([B(age) for age in X_te[:, 0]])

    # Append to design matrix
    X_tr = np.hstack([X_tr, Phi_tr])
    X_te = np.hstack([X_te, Phi_te])

    # Save expanded covariates
    cov_tr_path = os.path.join(roi_dir, 'cov_bspline_tr.txt')
    cov_te_path = os.path.join(roi_dir, 'cov_bspline_te.txt')
    np.savetxt(cov_tr_path, X_tr)
    np.savetxt(cov_te_path, X_te)

    # 9d) Prepare Response Paths
    resp_tr_path = os.path.join(output_dir, f'resp_tr_{roi}.txt')
    resp_te_path = os.path.join(output_dir, f'resp_te_{roi}.txt')

    # Change directory for PCNToolkit safety
    os.chdir(roi_dir)

    try:
        # 9e) ESTIMATE MODEL (with Fix)
        # We explicitly pass cv_folds=5 to prevent the TypeError
        yhat_te, s2_te, nm, Z, metrics_te = estimate(
            cov_tr_path, resp_tr_path,
            testresp=resp_te_path, testcov=cov_te_path,
            alg='blr', optimizer='powell',
            cv_folds=5,  # CRITICAL FIX
            savemodel=True, saveoutput=False,
            standardize=False
        )

        # Save outputs
        np.savetxt(f'{roi}_yhat_te.txt', yhat_te)
        np.savetxt(f'{roi}_s2_te.txt', s2_te)
        np.savetxt(f'{roi}_Z.txt', Z)
        print(f"   • Saved outputs for {roi}")

        # Collect Prediction Data
        metrics_globmetrics.loc[len(metrics_globmetrics)] = [
            roi,
            yhat_te.flatten().tolist(),
            s2_te.flatten().tolist(),
            Z.flatten().tolist()
        ]

        # Collect Test Metrics
        blr_metrics.loc[len(blr_metrics)] = [
            roi,
            metrics_te['MSLL'][0], metrics_te['EXPV'][0],
            metrics_te['SMSE'][0], metrics_te['RMSE'][0],
            metrics_te['Rho'][0],  metrics_te['pRho'][0],
        ]

        # 9f) GLOBAL EVALUATION
        y_te = np.loadtxt(resp_te_path)[:, None]
        y_tr = np.loadtxt(resp_tr_path)[:, None]

        metricsglob = evaluate(
            y_te, yhat_te, s2_te,
            np.array([y_tr.mean()]), np.array([y_tr.std()])
        )

        blr_globmetrics.loc[len(blr_globmetrics)] = [
            roi, 'global',
            metricsglob['MSLL'][0], metricsglob['EXPV'][0],
            metricsglob['SMSE'][0], metricsglob['RMSE'][0],
            metricsglob['Rho'][0],  metricsglob['pRho'][0],
        ]

        # 9g) SITE-LEVEL EVALUATION
        for num, site_idx in enumerate(sites):
            # Safety check: skip empty sites
            if len(site_idx) == 0: continue
            
            mets = evaluate(
                y_te[site_idx], yhat_te[site_idx], s2_te[site_idx],
                np.array([y_tr.mean()]), np.array([y_tr.std()])
            )
            
            blr_site_metrics.loc[len(blr_site_metrics)] = [
                roi, site_names[num],
                mets['MSLL'][0], mets['EXPV'][0],
                mets['SMSE'][0], mets['RMSE'][0],
                mets['Rho'][0],  mets['pRho'][0],
            ]
            
    except Exception as e:
        print(f"   ❌ Error processing {roi}: {str(e)}")

# ===========================
# 10) SAVE FINAL RESULTS
# ===========================
os.chdir(output_dir) # Return to main output dir
blr_metrics.to_csv('blr_metrics_new.csv', index=False)
blr_site_metrics.to_csv('blr_site_metrics.csv', index=False)
blr_globmetrics.to_csv('blr_globmetrics_new.csv', index=False)
metrics_globmetrics.to_csv('blr_yhat_s2_Z.csv', index=False)

print("✅ Pipeline Finished. All metrics and Z-scores saved.")
```
