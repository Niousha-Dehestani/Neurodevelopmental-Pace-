# 07 — Normative Centile Plots

Generates normative centile curves and overlays harmonised observed data for all structural MRI ROIs.

**Inputs:**
- `final_test.csv` — healthy test data (from `03_train_test_split.ipynb`)
- Trained BLR models (from `04_normative_model_blr.ipynb`)
- `Z_all_timepoints_healthy.csv` (from `06_longitudinal_predict_healthy.ipynb`) — used for site harmonisation

**Pipeline:**
1. Harmonise real test data to reference site → `final_harmonized_results.csv`
2. Generate fake age-sweep covariates (sex-specific, N=100, ages 9–17, reference site)
3. Run `predict()` → smooth centile curves (ŷhat ± σ)
4. All-ROI centile grid figure
5. Per-ROI individual centile figures

**Sex:** Set `SEX = 1` for male, `SEX = 2` for female in the config cell.  
**Environment:** `normmodel310` (PCNToolkit ≥ 0.29)

```python
# ── CONFIG ──────────────────────────────────────────────────────────────────
# Trained model root — one sub-folder per ROI, each containing a Models/ folder
MODEL_DIR       = 'outputs/normative_model'

# Test data (output of 03_train_test_split.ipynb)
INPUT_CSV       = 'data/splits/final_test.csv'

# Z-scores from 06_longitudinal_predict_healthy.ipynb (used for harmonisation)
Z_SCORES_CSV    = 'outputs/longitudinal_healthy/Z_all_timepoints_healthy.csv'

# All outputs (harmonised CSV, figures) are written here
OUTPUT_DIR      = 'outputs/centile_plots'

# Sex for centile curves: 1 = Male, 2 = Female
SEX             = 1

# Fake age-sweep parameters
N_FAKE          = 100        # number of synthetic subjects
AGE_MIN         = 9.0
AGE_MAX         = 17.0
REFERENCE_SITE  = 2          # site_mri integer used as the reference (site_mri_2 = 1)
N_SITES         = 30

# B-spline — must match training
AGE_XMIN, AGE_XMAX, AGE_NKNOTS = 7, 18, 3
```

```python
import os
import math
import numpy as np
import pandas as pd
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
from pcntoolkit.normative import predict
from pcntoolkit.util.utils import create_bspline_basis

plt.rcParams['font.family'] = 'Arial'

SEX_LABEL = 'male' if SEX == 1 else 'female'

HARM_CSV    = os.path.join(OUTPUT_DIR, 'final_harmonized_results.csv')
FAKE_DIR    = os.path.join(OUTPUT_DIR, 'fake_covariates')
GRID_DIR    = os.path.join(OUTPUT_DIR, 'grid')
INDIV_DIR   = os.path.join(OUTPUT_DIR, f'per_roi_{SEX_LABEL}')

for d in [OUTPUT_DIR, FAKE_DIR, GRID_DIR, INDIV_DIR]:
    os.makedirs(d, exist_ok=True)
```

## Step 1 — Load test data & identify ROIs

```python
DROP_COLS = ['mr_y_smri__vol__dsk__lh_sum', 'mr_y_smri__vol__dsk__rh_sum', 'mr_y_smri__vol__dsk_sum']

df_test = pd.read_csv(INPUT_CSV)
df_test = df_test.drop(columns=[c for c in DROP_COLS if c in df_test.columns])

roi_ids = df_test.columns[df_test.columns.str.contains('mr_y_smri__')].tolist()
print(f'Test rows: {len(df_test)} | ROIs: {len(roi_ids)}')
```

## Step 2 — Harmonise real test data to reference site

Covariates use real ages and sexes but all rows are assigned `site_mri_{REFERENCE_SITE} = 1`.
Harmonised value: `harm_roi = (Z_score × √s²) + ŷhat`  
This removes inter-site variance so observed scatter is comparable to centile curves from the reference site.

```python
# Build covariates (real age + sex, one-hot site, force reference site)
df_harm = df_test[['ab_g_dyn__visit_age', 'ab_g_stc__cohort_sex']].copy()

site_cols = [f'site_mri_{i}' for i in range(1, N_SITES + 1)]
if 'site_mri' in df_test.columns:
    site_dummies = pd.get_dummies(df_test['site_mri'], prefix='site_mri')
    site_dummies = site_dummies.reindex(columns=site_cols, fill_value=0).astype(int)
else:
    site_dummies = pd.DataFrame(0, index=df_harm.index, columns=site_cols)

df_harm = pd.concat([df_harm, site_dummies], axis=1)
df_harm[f'site_mri_{REFERENCE_SITE}'] = 1  # force reference site

cov_harm_path = os.path.join(FAKE_DIR, 'cov_harm.txt')
df_harm.to_csv(cov_harm_path, sep=' ', header=False, index=False)

# Save per-ROI response files
for roi in roi_ids:
    df_test[roi].to_csv(os.path.join(FAKE_DIR, f'resp_te_{roi}.txt'), header=False, index=False)

# Load Z-scores for harmonisation (optional)
z_control = pd.read_csv(Z_SCORES_CSV) if os.path.exists(Z_SCORES_CSV) else None
if z_control is None:
    print('Z-scores not found — harmonised columns will be skipped.')

# B-spline basis
B = create_bspline_basis(AGE_XMIN, AGE_XMAX, nknots=AGE_NKNOTS)

root_dir = os.getcwd()

for roi in roi_ids:
    roi_dir = os.path.join(MODEL_DIR, roi)
    roi_out = os.path.join(FAKE_DIR, f'harm_{roi}')
    model_folder = os.path.join(roi_dir, 'Models')
    os.makedirs(roi_out, exist_ok=True)

    if not os.path.exists(model_folder):
        continue

    # B-spline expand
    X = np.loadtxt(cov_harm_path)
    X = np.hstack([X, np.ones((X.shape[0], 1))])
    X = np.hstack([X, np.vstack([B(a) for a in X[:, 0]])])
    bsp_path = os.path.join(roi_dir, 'cov_bspline_harm.txt')
    np.savetxt(bsp_path, X)

    os.chdir(roi_out)
    try:
        predict(
            covfile=bsp_path,
            respfile=os.path.join(FAKE_DIR, f'resp_te_{roi}.txt'),
            alg='blr',
            model_path=model_folder,
            inputsuffix='estimate',
            outputsuffix='harm',
            return_y=False,
        )
        yhat_f = [f for f in os.listdir('.') if 'yhat' in f and 'harm' in f]
        ys2_f  = [f for f in os.listdir('.') if 'ys2'  in f and 'harm' in f]
        if yhat_f and ys2_f:
            yhat = pd.read_csv(yhat_f[0], header=None)[0].values
            ys2  = pd.read_csv(ys2_f[0],  header=None)[0].values
            df_harm[f'yhat_{roi}'] = yhat
            df_harm[f's2_{roi}']   = ys2
            if z_control is not None and roi in z_control.columns:
                df_harm[f'harm_{roi}'] = (z_control[roi].values * np.sqrt(ys2)) + yhat
    except Exception as e:
        print(f'  ERROR {roi}: {e}')

os.chdir(root_dir)
df_harm.to_csv(HARM_CSV, index=False)
print(f'Harmonised results saved: {HARM_CSV}')
```

## Step 3 — Split harmonised data by sex

```python
finalharm = pd.read_csv(HARM_CSV)
finalharm_ma = finalharm[finalharm['ab_g_stc__cohort_sex'] == 1]
finalharm_fe = finalharm[finalharm['ab_g_stc__cohort_sex'] == 2]
obs_data = finalharm_ma if SEX == 1 else finalharm_fe

print(f'Harmonised: males={len(finalharm_ma)}, females={len(finalharm_fe)}')
print(f'Using sex={SEX} ({SEX_LABEL}) for centile curves and observed scatter.')
```

## Step 4 — Generate fake age-sweep covariates

```python
np.random.seed(42)

site_cols = [f'site_mri_{i}' for i in range(1, N_SITES + 1)]
fake_data = {
    'ab_g_dyn__visit_age': np.random.uniform(AGE_MIN, AGE_MAX, N_FAKE),
    'ab_g_stc__cohort_sex': np.full(N_FAKE, SEX),
}
for c in site_cols:
    fake_data[c] = 1 if c == f'site_mri_{REFERENCE_SITE}' else 0

df_fake = pd.DataFrame(fake_data)
df_fake = df_fake.apply(pd.to_numeric, errors='coerce')

cov_fake_path = os.path.join(FAKE_DIR, 'cov_fake.txt')
df_fake.to_csv(cov_fake_path, sep=' ', header=False, index=False)

# Dummy response files (random, not used in evaluation)
for roi in roi_ids:
    np.savetxt(os.path.join(FAKE_DIR, f'resp_fake_{roi}.txt'), np.random.randn(N_FAKE, 1))

print(f'Fake covariates: {df_fake.shape}  sex={SEX} ({SEX_LABEL})  site=site_mri_{REFERENCE_SITE}')
```

## Step 5 — B-spline expansion & predict()

```python
root_dir = os.getcwd()

for roi in roi_ids:
    roi_dir      = os.path.join(MODEL_DIR, roi)
    roi_out      = os.path.join(FAKE_DIR, f'fake_{roi}')
    model_folder = os.path.join(roi_dir, 'Models')
    os.makedirs(roi_out, exist_ok=True)
    os.makedirs(os.path.join(roi_dir, 'blr'), exist_ok=True)

    if not os.path.exists(model_folder):
        continue

    # B-spline expand
    X = np.loadtxt(cov_fake_path)
    X = np.hstack([X, np.ones((X.shape[0], 1))])
    X = np.hstack([X, np.vstack([B(a) for a in X[:, 0]])])
    bsp_path = os.path.join(roi_dir, 'cov_bspline_fake.txt')
    np.savetxt(bsp_path, X)

    os.chdir(roi_out)
    try:
        predict(
            covfile=bsp_path,
            respfile=os.path.join(FAKE_DIR, f'resp_fake_{roi}.txt'),
            alg='blr',
            model_path=model_folder,
            inputsuffix='estimate',
            outputsuffix='totaltest',
            return_y=True,
        )
        yhat_files = [f for f in os.listdir('.') if 'yhat' in f and 'totaltest' in f]
        ys2_files  = [f for f in os.listdir('.') if 'ys2'  in f and 'totaltest' in f]
        if yhat_files and ys2_files:
            yhat = pd.read_csv(yhat_files[0], sep=' ', header=None).values.flatten()
            ys2  = pd.read_csv(ys2_files[0],  sep=' ', header=None).values.flatten()
            df_fake[f'yhat_{roi}'] = yhat
            df_fake[f's2_{roi}']   = ys2
        else:
            print(f'  MISSING output: {roi}')
    except Exception as e:
        print(f'  ERROR {roi}: {e}')

os.chdir(root_dir)
rois_ready = [r for r in roi_ids if f'yhat_{r}' in df_fake.columns]
print(f'Centile curves ready for {len(rois_ready)} / {len(roi_ids)} ROIs.')
```

## Step 6 — All-ROI centile grid

```python
n_cols = 8
n_rows = math.ceil(len(rois_ready) / n_cols)
fig, axes = plt.subplots(n_rows, n_cols, figsize=(40, n_rows * 5))
axes = axes.flatten()

df_plot = df_fake.sort_values('ab_g_dyn__visit_age')

for i, roi in enumerate(rois_ready):
    ax = axes[i]
    age   = df_plot['ab_g_dyn__visit_age']
    yhat  = df_plot[f'yhat_{roi}']
    sigma = np.sqrt(df_plot[f's2_{roi}'])

    ax.fill_between(age, yhat - 1.96*sigma, yhat + 1.96*sigma, color='steelblue', alpha=0.15)
    ax.fill_between(age, yhat - sigma,       yhat + sigma,       color='steelblue', alpha=0.25)
    ax.plot(age, yhat,                color='steelblue', lw=3)
    ax.plot(age, yhat + 1.96*sigma,   color='steelblue', lw=1, ls='--', alpha=0.5)
    ax.plot(age, yhat - 1.96*sigma,   color='steelblue', lw=1, ls='--', alpha=0.5)

    harm_col = f'harm_{roi}'
    if harm_col in obs_data.columns:
        ax.scatter(obs_data['ab_g_dyn__visit_age'], obs_data[harm_col],
                   color='black', alpha=0.4, s=15, edgecolors='none')

    ax.set_title(VOL_MAPPING.get(roi, roi), fontsize=22, fontweight='bold')
    ax.set_xlabel('Age (years)', fontsize=20)
    ax.set_ylabel('Volume (mm³)', fontsize=20)
    ax.tick_params(axis='both', labelsize=24)
    ax.grid(True, linestyle=':', alpha=0.3)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.tight_layout()
grid_path = os.path.join(GRID_DIR, f'centile_grid_{SEX_LABEL}.png')
plt.savefig(grid_path, dpi=300, bbox_inches='tight')
plt.close()
print(f'Saved: {grid_path}')
```

## Step 7 — Per-ROI individual centile figures

```python
df_plot = df_fake.sort_values('ab_g_dyn__visit_age')

for roi in rois_ready:
    try:
        age   = df_plot['ab_g_dyn__visit_age']
        yhat  = df_plot[f'yhat_{roi}']
        sigma = np.sqrt(df_plot[f's2_{roi}'])

        fig, ax = plt.subplots(figsize=(10, 6))

        # Shaded uncertainty bands
        ax.fill_between(age, yhat - 1.96*sigma, yhat + 1.96*sigma,
                        color='steelblue', alpha=0.10, label='2.5–97.5th centile')
        ax.fill_between(age, yhat - sigma,       yhat + sigma,
                        color='steelblue', alpha=0.20, label='16–84th centile')

        # Centile lines
        ax.plot(age, yhat,              color='steelblue', lw=2.5, label='50th centile')
        ax.plot(age, yhat + 1.96*sigma, color='steelblue', lw=0.5, ls='--', alpha=0.4)
        ax.plot(age, yhat - 1.96*sigma, color='steelblue', lw=0.5, ls='--', alpha=0.4)

        # Observed scatter (harmonised)
        harm_col = f'harm_{roi}'
        if harm_col in obs_data.columns:
            ax.scatter(obs_data['ab_g_dyn__visit_age'], obs_data[harm_col],
                       color='gray', alpha=0.15, s=12, edgecolors='none',
                       label=f'Observed ({SEX_LABEL})')

        clean_title = VOL_MAPPING.get(roi, roi)
        ax.set_title(f'Normative centiles: {clean_title}', fontsize=14, fontweight='bold')
        ax.set_xlabel('Age (years)', fontsize=11)
        ax.set_ylabel('Volume (mm³)', fontsize=11)
        ax.legend(frameon=False, loc='upper right', fontsize=9)
        ax.grid(True, linestyle=':', alpha=0.4)
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)

        plt.tight_layout()
        plt.savefig(os.path.join(INDIV_DIR, f'centile_{roi}_{SEX_LABEL}.png'),
                    dpi=300, bbox_inches='tight')
        plt.close()

    except Exception as e:
        print(f'  ERROR {roi}: {e}')

print(f'Per-ROI figures saved to: {INDIV_DIR}')
print('Pipeline complete.')
```
