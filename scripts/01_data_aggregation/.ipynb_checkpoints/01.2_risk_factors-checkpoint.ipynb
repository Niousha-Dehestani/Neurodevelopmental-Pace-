# Environmental Risk Factors — ABCD Study (Release 6.0)

**Purpose:** Load, scrub, and prepare environmental risk factor variables for normative modelling.

**Pipeline steps**
1. Load baseline phenotype parquet files and audit invalid codes  
2. Process 15 variable categories (demographics, birth history, BMI, sleep, lifestyle, social environment, puberty)  
3. Rename columns to human-readable labels  
4. Save merged risk factor CSV  
5. Plot variable distributions by category  

**Required ABCD Release 6.0 phenotype files** (set `PHENOTYPE_DIR` below):
```
ab_g_dyn.parquet     ab_g_stc.parquet     ab_p_demo.parquet
ph_p_dhx.parquet     mh_p_famhx.parquet   ph_y_anthr.parquet
ph_p_sds.parquet     nt_p_yst.parquet     ph_y_pa.parquet
mh_p_kbi.parquet     fc_y_fes.parquet     fc_p_nsc.parquet
fc_y_pm.parquet      fc_y_crpbi.parquet   fc_p_psb.parquet
fc_y_srpf.parquet    ph_p_pds.parquet
```

## Helper — invalid code scrubber
Replaces ABCD placeholder codes (999, 888, 777, 555, 444) with `NaN`.

````python
INVALID_CODES = [999, 888, 777, 555, 444, "999", "888", "777", "555", "444"]

def scrub(df):
    return df.astype(object).replace(INVALID_CODES, np.nan).infer_objects(copy=False)

def load_baseline(filename):
    df = scrub(pd.read_parquet(PHENOTYPE_DIR / filename))
    if "session_id" in df.columns:
        df = df[df["session_id"] == "ses-00A"]
    return df.set_index("participant_id")
````

## Step 1 — Variable audit (optional QC)

Checks every target variable for missingness and ABCD invalid-code frequency.  
Saves `variable_audit_report.csv` to `OUTPUT_DIR`.

````python
AUDIT_ITEMS = [
    "ab_g_dyn__cohort_edu__cgs", "ab_g_dyn__design_mr__serial", "ab_g_dyn__visit_age",
    "ab_g_dyn__design_site", "ab_g_stc__design_id__fam", "ab_g_stc__cohort_ethn",
    "ab_g_stc__cohort_race__nih", "ab_g_stc__cohort_sex", "ab_p_demo__income__hhold_001",
    "ph_p_dhx_003__01", "ph_p_dhx_004__01", "ph_p_dhx_006",
    "ph_p_dhx_012", "ph_p_dhx_013", "ph_p_dhx_014", "ph_p_dhx__birth_010",
    "ph_p_dhx__med_010", "ph_p_dhx__med_011", "ph_p_dhx__med_012",
    "ph_p_dhx__alc_001a", "ph_p_dhx__alc_001b", "ph_p_dhx__coc_001a", "ph_p_dhx__coc_001b",
    "ph_p_dhx__mj_001a", "ph_p_dhx__mj_001b", "ph_p_dhx__nic_001a", "ph_p_dhx__nic_001b",
    "ph_p_dhx__opi_001a", "ph_p_dhx__opi_001b", "ph_p_dhx__vit_001",
    "mh_p_famhx__dep_001", "mh_p_famhx__alc_001", "mh_p_famhx__drg_001",
    "mh_p_famhx__halluc_001", "mh_p_famhx__mania_001", "mh_p_famhx__nerve_001",
    "mh_p_famhx__suic_001", "mh_p_famhx__troub_001",
    "ph_y_anthr__height__r01_001", "ph_y_anthr__height__r02_001", "ph_y_anthr__height__r03_001",
    "ph_y_anthr__weight__r01_001", "ph_y_anthr__weight__r02_001", "ph_y_anthr__weight__r03_001",
    "ph_p_sds__da_001", "ph_p_sds__da_002", "ph_p_sds__da_003",
    "ph_p_sds__dims_001", "ph_p_sds__dims_002", "ph_p_sds__dims_003", "ph_p_sds__dims_004",
    "ph_p_sds__dims_005", "ph_p_sds__dims_006", "ph_p_sds__dims_007",
    "ph_p_sds__does_001", "ph_p_sds__does_002", "ph_p_sds__does_003",
    "ph_p_sds__does_004", "ph_p_sds__does_005",
    "ph_p_sds__hyphy_001", "ph_p_sds__hyphy_002",
    "ph_p_sds__sbd_001", "ph_p_sds__sbd_002", "ph_p_sds__sbd_003",
    "ph_p_sds__swtd_001", "ph_p_sds__swtd_002", "ph_p_sds__swtd_003",
    "ph_p_sds__swtd_004", "ph_p_sds__swtd_005", "ph_p_sds__swtd_006",
    "nt_p_yst__screen__wkdy_sum", "nt_p_yst__screen__wknd_sum",
    "ph_y_pa_001", "ph_y_pa_002", "ph_y_pa_003",
    "fc_p_nsc__ns_001", "fc_p_nsc__ns_002", "fc_p_nsc__ns_003", "fc_y_pm_mean",
    "mh_p_kbi__bully_001", "fc_y_fes__confl_mean", "fc_y_crpbi__cg1_mean",
    "fc_p_psb_mean", "fc_y_srpf__dis_mean", "fc_y_srpf__env_mean", "fc_y_srpf__involv_mean",
    "ph_p_pds__m_mean", "ph_p_pds__f_mean",
]

# Load all source files and concatenate
filenames = [
    "ab_g_dyn.parquet", "ab_g_stc.parquet", "ab_p_demo.parquet",
    "ph_p_dhx.parquet", "mh_p_famhx.parquet", "ph_y_anthr.parquet",
    "ph_p_sds.parquet", "nt_p_yst.parquet", "ph_y_pa.parquet",
    "mh_p_kbi.parquet", "fc_y_fes.parquet", "fc_p_nsc.parquet",
    "fc_y_pm.parquet", "fc_y_crpbi.parquet", "fc_p_psb.parquet",
    "fc_y_srpf.parquet", "ph_p_pds.parquet",
]

all_dfs = [load_baseline(f) for f in filenames]
raw = pd.concat([d for d in all_dfs if not d.empty], axis=1)

found     = [c for c in AUDIT_ITEMS if c in raw.columns]
not_found = [c for c in AUDIT_ITEMS if c not in raw.columns]
if not_found:
    print(f"WARNING: {len(not_found)} variables not found in source files:")
    for v in not_found:
        print(f"  - {v}")

rows = []
for col in found:
    s = raw[col]
    row = {"Variable": col, "dtype": str(s.dtype),
           "N": len(s), "Missing": s.isna().sum(),
           "Invalid_codes": s.isin(INVALID_CODES).sum()}
    for code_ in [999, 888, 777, 555, 444]:
        row[f"n_{code_}"] = s.isin([code_, str(code_)]).sum()
    rows.append(row)

audit_df = pd.DataFrame(rows)
audit_path = OUTPUT_DIR / "variable_audit_report.csv"
audit_df.to_csv(audit_path, index=False)
print(f"Audit report saved: {audit_path}  ({len(audit_df)} variables)")
audit_df.head()
````

## Step 2 — Process all risk factor categories

Loads each phenotype file, scrubs invalid codes, computes composite scores,  
and concatenates everything into a single baseline dataframe.

````python
n_audit = {}

# ── 1. Demographics ────────────────────────────────────────────
dyn = load_baseline("ab_g_dyn.parquet")
dyn["scanner_site"] = (dyn["ab_g_dyn__design_mr__serial"].astype(str)
                       + "_" + dyn["ab_g_dyn__design_site"].astype(str))
dyn_c = dyn[["ab_g_dyn__visit_age", "ab_g_dyn__cohort_edu__cgs", "scanner_site"]].dropna()
n_audit["Demographics_Dyn"] = len(dyn_c)

stc = load_baseline("ab_g_stc.parquet")
stc_c = stc[["ab_g_stc__design_id__fam", "ab_g_stc__cohort_ethn",
             "ab_g_stc__cohort_race__nih", "ab_g_stc__cohort_sex"]].dropna()
n_audit["Demographics_Stc"] = len(stc_c)

pdemo = load_baseline("ab_p_demo.parquet")
demo_c = pdemo[["ab_p_demo__income__hhold_001"]].apply(pd.to_numeric, errors="coerce").dropna()
n_audit["Demographics_Parent"] = len(demo_c)

# ── 2. Birth History ───────────────────────────────────────────
birth_items = [
    "ph_p_dhx_003__01", "ph_p_dhx_004__01", "ph_p_dhx__birth_001", "ph_p_dhx__birth_003",
    "ph_p_dhx__birth_004", "ph_p_dhx__birth_005", "ph_p_dhx__birth_006", "ph_p_dhx__birth_007",
    "ph_p_dhx__birth_008", "ph_p_dhx__birth_009", "ph_p_dhx__birth_010", "ph_p_dhx_006",
    "ph_p_dhx__med_001", "ph_p_dhx__med_002", "ph_p_dhx__med_003", "ph_p_dhx__med_004",
    "ph_p_dhx__med_005", "ph_p_dhx__med_006", "ph_p_dhx__med_007", "ph_p_dhx__med_008",
    "ph_p_dhx__med_009", "ph_p_dhx__med_010", "ph_p_dhx__med_011", "ph_p_dhx__med_012",
    "ph_p_dhx__alc_001a", "ph_p_dhx__alc_001b", "ph_p_dhx__coc_001a", "ph_p_dhx__coc_001b",
    "ph_p_dhx__mj_001a", "ph_p_dhx__mj_001b", "ph_p_dhx__nic_001a", "ph_p_dhx__nic_001b",
    "ph_p_dhx__opi_001a", "ph_p_dhx__opi_001b", "ph_p_dhx__rxpain_001a", "ph_p_dhx__rxpain_001b",
    "ph_p_dhx__vit_001", "ph_p_dhx_012", "ph_p_dhx_013", "ph_p_dhx_014",
]
birth = load_baseline("ph_p_dhx.parquet")
birth_c = birth[[c for c in birth_items if c in birth.columns]].apply(pd.to_numeric, errors="coerce").dropna()
n_audit["Birth_History"] = len(birth_c)

# ── 3. Family Mental Health ────────────────────────────────────
famhx_items = ["mh_p_famhx__dep_001", "mh_p_famhx__alc_001", "mh_p_famhx__drg_001",
               "mh_p_famhx__halluc_001", "mh_p_famhx__mania_001", "mh_p_famhx__nerve_001",
               "mh_p_famhx__suic_001", "mh_p_famhx__troub_001"]
famhx = load_baseline("mh_p_famhx.parquet")
famhx_num = famhx[famhx_items].apply(pd.to_numeric, errors="coerce").dropna()
famhx_num["Family Mentalhealth"] = famhx_num.mean(axis=1)
famhx_c = famhx_num[["Family Mentalhealth"]]
n_audit["Family_MH"] = len(famhx_c)

# ── 4. BMI + Waist ─────────────────────────────────────────────
antho = load_baseline("ph_y_anthr.parquet").apply(pd.to_numeric, errors="coerce")
antho["H_cm"] = antho[["ph_y_anthr__height__r01_001",
                        "ph_y_anthr__height__r02_001",
                        "ph_y_anthr__height__r03_001"]].mean(axis=1) * 2.54
antho["W_kg"] = antho[["ph_y_anthr__weight__r01_001",
                        "ph_y_anthr__weight__r02_001",
                        "ph_y_anthr__weight__r03_001"]].mean(axis=1) * 0.453592
antho["BMI"] = antho["W_kg"] / (antho["H_cm"] / 100) ** 2
antho_c = antho[["ph_y_anthr__waist_001", "BMI"]].dropna()
n_audit["Anthropometrics"] = len(antho_c)

# ── 5. Sleep (SDSC) ────────────────────────────────────────────
sleep_items = {
    "sleep_da":   ["ph_p_sds__da_001",   "ph_p_sds__da_002",   "ph_p_sds__da_003"],
    "sleep_dims": ["ph_p_sds__dims_001", "ph_p_sds__dims_002", "ph_p_sds__dims_003",
                   "ph_p_sds__dims_004", "ph_p_sds__dims_005", "ph_p_sds__dims_006", "ph_p_sds__dims_007"],
    "sleep_does": ["ph_p_sds__does_001", "ph_p_sds__does_002", "ph_p_sds__does_003",
                   "ph_p_sds__does_004", "ph_p_sds__does_005"],
    "sleep_hyphy":["ph_p_sds__hyphy_001","ph_p_sds__hyphy_002"],
    "sleep_sbd":  ["ph_p_sds__sbd_001",  "ph_p_sds__sbd_002",  "ph_p_sds__sbd_003"],
    "sleep_swtd": ["ph_p_sds__swtd_001", "ph_p_sds__swtd_002", "ph_p_sds__swtd_003",
                   "ph_p_sds__swtd_004", "ph_p_sds__swtd_005", "ph_p_sds__swtd_006"],
}
all_sleep_cols = [c for cols in sleep_items.values() for c in cols]
sleep = load_baseline("ph_p_sds.parquet")
sleep_num = sleep[all_sleep_cols].apply(pd.to_numeric, errors="coerce").dropna()
for subscale, cols in sleep_items.items():
    sleep_num[subscale] = sleep_num[cols].mean(axis=1)
sleep_c = sleep_num[list(sleep_items.keys())]
n_audit["Sleep_SDSC"] = len(sleep_c)

# ── 6–14. Lifestyle & Social Environment ──────────────────────
screen = load_baseline("nt_p_yst.parquet")
screen_c = screen[["nt_p_yst__screen__wkdy_sum", "nt_p_yst__screen__wknd_sum"]].apply(pd.to_numeric, errors="coerce").dropna()
screen_c["screen_time"] = screen_c.mean(axis=1)
screen_c = screen_c[["screen_time"]]
n_audit["Screen_Time"] = len(screen_c)

pa = load_baseline("ph_y_pa.parquet")
pa_c = pa[["ph_y_pa_001", "ph_y_pa_002", "ph_y_pa_003"]].apply(pd.to_numeric, errors="coerce").dropna()
pa_c["ph_y_mean"] = -pa_c.mean(axis=1)
pa_c = pa_c[["ph_y_mean"]]
n_audit["Physical_Activity"] = len(pa_c)

kbi_c = load_baseline("mh_p_kbi.parquet")[["mh_p_kbi__bully_001"]].dropna()
n_audit["Bullying"] = len(kbi_c)

fes_c = load_baseline("fc_y_fes.parquet")[["fc_y_fes__confl_mean"]].dropna()
n_audit["Family_Conflict"] = len(fes_c)

nsc = load_baseline("fc_p_nsc.parquet")
nsc_c = nsc[["fc_p_nsc__ns_001", "fc_p_nsc__ns_002", "fc_p_nsc__ns_003"]].apply(pd.to_numeric, errors="coerce").dropna()
nsc_c["nsc_mean"] = -nsc_c.mean(axis=1)
nsc_c = nsc_c[["nsc_mean"]]
n_audit["Neighbourhood_Safety"] = len(nsc_c)

pm_c = load_baseline("fc_y_pm.parquet")[["fc_y_pm_mean"]].dropna()
pm_c["fc_y_pm_mean"] = -pm_c["fc_y_pm_mean"]
n_audit["Physical_Neglect"] = len(pm_c)

crpbi_c = load_baseline("fc_y_crpbi.parquet")[["fc_y_crpbi__cg1_mean"]].dropna()
crpbi_c["fc_y_crpbi__cg1_mean"] = -crpbi_c["fc_y_crpbi__cg1_mean"]
n_audit["Emotional_Neglect"] = len(crpbi_c)

psb_c = load_baseline("fc_p_psb.parquet")[["fc_p_psb_mean"]].dropna()
psb_c["fc_p_psb_mean"] = -psb_c["fc_p_psb_mean"]
n_audit["Prosocial"] = len(psb_c)

srpf = load_baseline("fc_y_srpf.parquet")
srpf_c = srpf[["fc_y_srpf__dis_mean", "fc_y_srpf__env_mean", "fc_y_srpf__involv_mean"]].dropna()
srpf_c["fc_y_srpf__env_mean"]    = -srpf_c["fc_y_srpf__env_mean"]
srpf_c["fc_y_srpf__involv_mean"] = -srpf_c["fc_y_srpf__involv_mean"]
n_audit["School"] = len(srpf_c)

# ── 15. Puberty (PDS) ──────────────────────────────────────────
pds = scrub(pd.read_parquet(PHENOTYPE_DIR / "ph_p_pds.parquet"))
pds = pds[pds["session_id"] == "ses-00A"]
female_pds = pds[["participant_id", "ph_p_pds__f_mean"]].dropna().rename(columns={"ph_p_pds__f_mean": "pds_total"}).set_index("participant_id")
male_pds   = pds[["participant_id", "ph_p_pds__m_mean"]].dropna().rename(columns={"ph_p_pds__m_mean": "pds_total"}).set_index("participant_id")
pubrty_c   = pd.concat([male_pds, female_pds], axis=0)
n_audit["Puberty_PDS"] = len(pubrty_c)

# ── Final merge (inner join) ───────────────────────────────────
env_c = pd.concat([
    pubrty_c, srpf_c, psb_c, crpbi_c, pm_c, nsc_c, fes_c,
    pa_c, kbi_c, screen_c, famhx_c, sleep_c, birth_c,
    demo_c, antho_c, stc_c, dyn_c,
], axis=1, join="inner")

print("=" * 45)
print("SAMPLE SIZE AUDIT")
print("=" * 45)
for cat, n in n_audit.items():
    print(f"  {cat:<25}: {n:,}")
print("-" * 45)
print(f"  Final N (inner join)     : {len(env_c):,}")
print(f"  Columns                  : {env_c.shape[1]}")

out_path = OUTPUT_DIR / "env_risk_factors_raw.csv"
env_c.to_csv(out_path)
print(f"\nSaved: {out_path}")
````

## Step 4 — Variable distribution plots

One subplot per variable, coloured by category.  
Binary variables shown as bar plots; continuous as KDE histograms.  
Saved to `OUTPUT_DIR/figures/`.

````python
CATEGORY_COLORS = {
    "Prenatal/Birth History": "royalblue",
    "Infancy Development":    "green",
    "Biological Factors":     "red",
    "Lifestyle":              "darkorchid",
    "Social Environments":    "orange",
}

GROUP_MAP = {
    "Prenatal/Birth History": [
        "Maternal Age", "Paternal Age", "Prematurity", "Cyanosis at Birth",
        "Bradycardia at Birth", "Apnea at Birth", "Neonatal Convulsions",
        "Jaundice Treatment", "Oxygen Required", "Blood Transfusion",
        "Rh Incompatibility", "Planned Pregnancy", "Nausea/Vomiting",
        "Heavy Bleeding", "Pre-eclampsia", "Gallbladder Attack", "Proteinuria",
        "Rubella", "Anemia", "Infection", "Diabetes", "Hypertension",
        "Placental Complications", "Accident/Injury",
        "Alcohol(Pre)", "Alcohol(During)", "Cocaine(Pre)", "Cocaine(During)",
        "Marijuana(Pre)", "Marijuana(During)", "Tobacco(Pre)", "Tobacco(During)",
        "Opioids(Pre)", "Opioids(During)", "Rx Opioids(Pre)", "Rx Opioids(During)",
        "Prenatal Vitamins",
    ],
    "Infancy Development": [
        "Breastfeeding Duration", "Motor Development", "Speech Development",
    ],
    "Biological Factors": [
        "Pubertal development", "Waist", "BMI",
    ],
    "Lifestyle": [
        "Physical Activity", "Screen Time",
        "Arousal Disorder", "Sleep Initiation/Maintenance", "Excessive Sleepiness",
        "Sleep Hyperhidrosis", "Sleep Breathing Disorder", "Sleep-Wake Transition",
    ],
    "Social Environments": [
        "School Disengagement", "School Environment", "School Involvement",
        "Prosocial Behavior", "Emotional Neglect", "Physical Neglect",
        "Neighborhood Safety", "Family Conflict", "Family Mentalhealth",
        "Bullying", "Income(P)", "Education(P)",
    ],
}

def plot_distributions(df, group_map, category_colors, filename):
    ordered = [(var, cat)
               for cat, vars_ in group_map.items()
               for var in vars_
               if var in df.columns]
    if not ordered:
        print("No matching variables found.")
        return

    age_vars = {"Maternal Age", "Paternal Age"}
    n_vars, cols_per_row = len(ordered), 5
    n_rows = math.ceil(n_vars / cols_per_row)

    fig, axes = plt.subplots(n_rows, cols_per_row,
                             figsize=(50, n_rows * 5))
    axes_list = axes.flatten() if n_vars > 1 else [axes]

    for i, (col, cat) in enumerate(ordered):
        ax = axes_list[i]
        color = category_colors.get(cat, "gray")
        unique_vals = df[col].dropna().unique()
        is_binary = set(unique_vals).issubset({0, 1})

        if is_binary:
            sns.histplot(df[col], ax=ax, color=color, edgecolor="white",
                         discrete=True, alpha=0.8, shrink=0.4)
            ax.set_xticks([0, 1])
            ax.set_xlim(-0.5, 1.5)
        else:
            sns.histplot(df[col], kde=True, ax=ax, color=color,
                         edgecolor="white", alpha=0.7)
            if col in age_vars:
                ax.set_xlim(0, 100)
                ax.set_xticks(range(0, 101, 20))

        ax.set_title(col,  fontsize=42, fontweight="bold")
        ax.set_xlabel("",  fontsize=42)
        ax.set_ylabel("Frequency", fontsize=42, fontweight="bold")
        ax.tick_params(axis="both", labelsize=38)

    for j in range(i + 1, len(axes_list)):
        fig.delaxes(axes_list[j])

    plt.tight_layout()
    path = FIGURES_DIR / filename
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {path}")

plot_distributions(env, GROUP_MAP, CATEGORY_COLORS, "risk_factor_distributions.png")
````

## Step 5 — Save category legend separately

````python
def save_legend(category_colors, filename):
    patches = [mpatches.Patch(color=c, label=l)
               for l, c in category_colors.items()]
    fig = plt.figure(figsize=(12, 4))
    fig.legend(handles=patches, loc="center", ncol=2,
               columnspacing=1.5, fontsize=35,
               title="Variable Categories", title_fontsize=38,
               frameon=True, edgecolor="black")
    plt.axis("off")
    path = FIGURES_DIR / filename
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {path}")

save_legend(CATEGORY_COLORS, "risk_factor_legend.png")
````
