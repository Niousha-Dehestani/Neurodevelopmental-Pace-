# Train / Test Split — ABCD Study (Release 6.0)

**Purpose:** Split the QC-passed T1 dataset into normative training and held-out test sets.

**Pipeline steps**
1. Identify healthy vs clinical participants using medical screening (`ab_p_screen`)  
2. Load T1 MRI features + covariates (age, sex, site, race/ethnicity, family ID)  
3. Longitudinal 70/30 subject-level split of healthy participants  
4. Familial deduplication — one scan per family for each training pool  
5. Site filtering — remove sites with N < 10 in training set  
6. Apply site label mapping and save four output CSVs  
7. Visualisations: site age distributions, typical vs atypical comparison, individual trajectories  

**Inputs** (set paths in the Configuration cell):
- `T1_PROCESSED_CSV` — output of `01_t1_qc.ipynb`
- `PHENOTYPE_DIR`    — ABCD Release 6.0 phenotype folder

**Outputs** saved to `OUTPUT_DIR`:
```
normative_train.csv      ← one scan/family, normative training pool
normative_train2.csv     ← second independent scan/family pool
healthy_test.csv         ← 30% held-out typical longitudinal subjects
clinical_test.csv        ← all clinical subjects (ab_p_screen positives)
screen_psycho_summary.csv← screening diagnosis categories
```


## Configuration

````python
from pathlib import Path

# ── Set these paths ─────────────────────────────────────────────
PHENOTYPE_DIR    = Path("path/to/abcd6.0/phenotype")
T1_PROCESSED_CSV = Path("path/to/output/T1_processed.csv")   # from 01_t1_qc.ipynb
OUTPUT_DIR       = Path("output")
# ────────────────────────────────────────────────────────────────

FIGURES_DIR = OUTPUT_DIR / "figures"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

TEST_SIZE    = 0.30
RANDOM_STATE = 42
MIN_SITE_N   = 10      # drop sites with fewer participants in training set
````

## Imports

````python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.cm import ScalarMappable
from matplotlib.colors import Normalize
import seaborn as sns
import colorcet as cc
from sklearn.model_selection import train_test_split

mpl.use("Agg")
mpl.rcParams.update({
    "font.family":     "sans-serif",
    "font.sans-serif": ["Arial"],
    "savefig.dpi":     400,
})
print("Imports OK")
````

## Step 1 — Medical screening: healthy vs clinical

Participants with any positive response on neurological/psychiatric screening  
(`ab_p_screen__med_002` through `_017`) are classified as clinical.  
All others form the healthy reference pool.

````python
MED_COLS = [
    "ab_p_screen__med_002", "ab_p_screen__med_003", "ab_p_screen__med_004",
    "ab_p_screen__med_005", "ab_p_screen__med_006", "ab_p_screen__med_007",
    "ab_p_screen__med_008", "ab_p_screen__med_011", "ab_p_screen__med_012",
    "ab_p_screen__med_013", "ab_p_screen__med_015", "ab_p_screen__med_016",
    "ab_p_screen__med_017",
]

DX_MAPPING = {
    "ab_p_screen__med_016": "intellectual_disability",
    "ab_p_screen__med_015": "substance_disorder",
    "ab_p_screen__med_013": "autism",
    "ab_p_screen__med_012": "schizophrenia",
    "ab_p_screen__med_011": "multiple_sclerosis",
    "ab_p_screen__med_008": "epilepsy",
    "ab_p_screen__med_007": "subdural_haematoma",
    "ab_p_screen__med_006": "haemorrhage",
    "ab_p_screen__med_005": "aneurysm",
    "ab_p_screen__med_004": "stroke",
    "ab_p_screen__med_003": "brain_tumour",
    "ab_p_screen__med_002": "cerebral_palsy",
    "ab_p_screen__med_017": "any_psychosis",
}

screen = pd.read_parquet(PHENOTYPE_DIR / "ab_p_screen.parquet").set_index("participant_id")

# Label diagnosis category (first positive match wins)
screen["dxcat"] = pd.NA
for col, label in DX_MAPPING.items():
    if col in screen.columns:
        screen.loc[screen[col].astype(str) == "1", "dxcat"] = label

is_clinical  = (screen[MED_COLS].astype(str) == "1").any(axis=1)
clinical_ids = screen[is_clinical].index
healthy_ids  = screen[~is_clinical].index

screen[is_clinical].to_csv(OUTPUT_DIR / "screen_psycho_summary.csv")

print(f"Healthy participants:  {len(healthy_ids):,}")
print(f"Clinical participants: {len(clinical_ids):,}")
print(f"Diagnosis breakdown:")
print(screen.loc[is_clinical, "dxcat"].value_counts().to_string())
````

## Step 2 — Load T1 features and merge covariates

Loads QC-passed T1 data, then joins age, sex, site, race/ethnicity, and family ID  
from the dynamic (`ab_g_dyn`) and static (`ab_g_stc`) phenotype tables.

````python
# ── T1 MRI (from 01_t1_qc.ipynb output) ──────────────────────
t1 = pd.read_csv(T1_PROCESSED_CSV).set_index("ID-wave")
t1["participant_id"] = t1.index.str.slice(stop=12)
t1["session_id"]     = t1.index.str.slice(start=13)

# ── Dynamic covariates: age, site, MRI serial ─────────────────
dyn = pd.read_parquet(PHENOTYPE_DIR / "ab_g_dyn.parquet")
dyn["ID-wave"] = dyn["participant_id"].astype(str) + "_" + dyn["session_id"].astype(str)
dyn = dyn.set_index("ID-wave")[[
    "participant_id", "ab_g_dyn__visit_age",
    "ab_g_dyn__design_site", "ab_g_dyn__design_mr__serial",
]]
dyn["site_mri"] = (dyn["ab_g_dyn__design_mr__serial"].astype(str)
                   + "_" + dyn["ab_g_dyn__design_site"].astype(str))

# ── Static covariates: sex, family ID, race/ethnicity ─────────
stc = pd.read_parquet(PHENOTYPE_DIR / "ab_g_stc.parquet").set_index("participant_id")
stc_cols = ["ab_g_stc__cohort_sex", "ab_g_stc__design_id__fam",
            "ab_g_stc__cohort_ethn", "ab_g_stc__cohort_race__nih"]

covs = dyn.join(stc[stc_cols], on="participant_id", how="left")
covs_clean = covs.dropna(subset=["ab_g_stc__cohort_race__nih", "ab_g_stc__cohort_ethn"]).copy()

# ── Merge T1 + covariates, split by screening ─────────────────
def merge_cohort(id_set):
    subset = t1[t1["participant_id"].isin(id_set)]
    merged = pd.concat([covs_clean, subset], axis=1, join="inner")
    return merged.loc[:, ~merged.columns.duplicated()].dropna()

healthy_df   = merge_cohort(healthy_ids)
clinical_df  = merge_cohort(clinical_ids)

print(f"Healthy scans:  {len(healthy_df):,}  ({healthy_df['participant_id'].nunique():,} subjects)")
print(f"Clinical scans: {len(clinical_df):,}  ({clinical_df['participant_id'].nunique():,} subjects)")
````

## Step 3 — Longitudinal 70/30 subject-level split

Only subjects with **both** baseline (`ses-00A`) and 2-year (`ses-02A`) scans  
are eligible for the held-out test set.  
The split is performed at the **subject level** to prevent data leakage.

````python
# Subjects with ≥ 2 sessions
session_counts  = healthy_df["participant_id"].value_counts()
longitudinal_ids = session_counts[session_counts >= 2].index.tolist()
print(f"Longitudinal healthy subjects: {len(longitudinal_ids):,}")

train_ids, test_ids = train_test_split(
    longitudinal_ids, test_size=TEST_SIZE, random_state=RANDOM_STATE
)
train_ids, test_ids = set(train_ids), set(test_ids)

healthy_test = healthy_df[healthy_df["participant_id"].isin(test_ids)].copy()
train_pool   = healthy_df[~healthy_df["participant_id"].isin(test_ids)].copy()

print(f"Train pool subjects: {train_pool['participant_id'].nunique():,}  ({len(train_pool):,} scans)")
print(f"Test  set subjects:  {healthy_test['participant_id'].nunique():,}  ({len(healthy_test):,} scans)")
````

## Step 4 — Familial deduplication

Normative modelling requires independent observations.  
We sample **one scan per family** from the training pool to create two independent sets:
- `normative_train` — primary training set (one scan/family)
- `normative_train2` — secondary independent set (different sibling/scan)

````python
FAM_COL = "ab_g_stc__design_id__fam"

train1 = train_pool.groupby(FAM_COL, group_keys=False).sample(n=1, random_state=RANDOM_STATE)

remainder = train_pool[~train_pool.index.isin(train1.index)]
train2 = remainder.groupby(FAM_COL, group_keys=False).sample(n=1, random_state=RANDOM_STATE)

print(f"Normative train 1: {len(train1):,} scans ({train1[FAM_COL].nunique():,} families)")
print(f"Normative train 2: {len(train2):,} scans ({train2[FAM_COL].nunique():,} families)")
````

## Step 5 — Site filtering and label mapping

Sites with fewer than `MIN_SITE_N` (10) participants in `normative_train` are excluded  
from all splits to ensure stable normative model estimation per site.

````python
SITE_LABEL_MAP = {
    "HASH6b4422a7_1":  "site1:#1",  "HASH1314a204_2":  "site2:#1",
    "HASH5b0cf1bb_3":  "site3:#1",  "HASH4b0b8b05_4":  "site4:#1",
    "HASHfeb7e81a_4":  "site4:#2",  "HASH311170b9_5":  "site5:#1",
    "HASH96a0c182_6":  "site6:#1",  "HASH5ac2b20b_6":  "site6:#2",
    "HASH65b39280_7":  "site7:#1",  "HASH5b2fcf80_8":  "site8:#1",
    "HASH4d1ed7b1_9":  "site9:#1",  "HASH4b0b8b05_9":  "site9:#2",
    "HASHd7cb4c6d_10": "site10:#1", "HASHe3ce02d3_10": "site10:#2",
    "HASH03db707f_11": "site11:#1", "HASH31ce566d_12": "site12:#1",
    "HASHe4f6957a_12": "site12:#2", "HASHc3bf3d9c_13": "site13:#1",
    "HASH69f406fa_13": "site13:#2", "HASHc1b3365b_13": "site13:#3",
    "HASH7f91147d_14": "site14:#1", "HASH11ad4ed5_14": "site14:#2",
    "HASH7911780b_15": "site15:#1", "HASH96a0c182_16": "site16:#1",
    "HASH3935c89e_16": "site16:#2", "HASHdb2589d4_17": "site17:#1",
    "HASHa3e45734_18": "site18:#1", "HASH3ce956bd_18": "site18:#2",
    "HASH5ac2b20b_19": "site19:#1", "HASHa2f57943_20": "site20:#1",
    "HASHc9398971_20": "site20:#2", "HASHd422be27_20": "site20:#3",
    "HASHb640a1b8_21": "site21:#1", "HASH4036a433_21": "site21:#2",
    "HASHe76e6d72_21": "site21:#3", "HASHba667216_21": "site21:#4",
    "HASH48f7cbc3_22": "site22:#1",
}

site_counts  = train1["site_mri"].value_counts()
sites_to_keep = site_counts[site_counts >= MIN_SITE_N].index
n_dropped = len(site_counts) - len(sites_to_keep)
print(f"Sites retained: {len(sites_to_keep)}  |  dropped (N < {MIN_SITE_N}): {n_dropped}")

def apply_site_filter(df):
    out = df[df["site_mri"].isin(sites_to_keep)].copy()
    out["site_mri_label"] = out["site_mri"].map(SITE_LABEL_MAP)
    return out

normative_train  = apply_site_filter(train1)
normative_train2 = apply_site_filter(train2)
healthy_test_f   = apply_site_filter(healthy_test)
clinical_test_f  = apply_site_filter(clinical_df)
````

## Step 6 — Save outputs

````python
def save(df, name):
    path = OUTPUT_DIR / name
    df.to_csv(path)
    print(f"  {name}: {df.shape[0]:,} scans × {df.shape[1]} cols")

print("Saving …")
save(normative_train,  "normative_train.csv")
save(normative_train2, "normative_train2.csv")
save(healthy_test_f,   "healthy_test.csv")
save(clinical_test_f,  "clinical_test.csv")
print("Done.")
````

## Plot 1 — Age distribution across MRI sites

Violin + white boxplot overlay for the normative training set.  
Violins are coloured by actual sample size (continuous colormap).

````python
AGE_COL  = "ab_g_dyn__visit_age"

site_counts = normative_train2["site_mri_label"].value_counts()
site_order  = site_counts.sort_values(ascending=False).index

norm = Normalize(vmin=site_counts.min(), vmax=site_counts.max())
cmap = cc.cm["CET_L8"]
palette = {site: cmap(norm(count)) for site, count in site_counts.items()}

TITLE_SIZE, LABEL_SIZE, TICK_SIZE = 30, 28, 26

fig, ax = plt.subplots(figsize=(20, 13))
sns.set_style("whitegrid", {"axes.grid": False})

sns.violinplot(
    data=normative_train2, x="site_mri_label", y=AGE_COL,
    hue="site_mri_label", order=site_order, palette=palette,
    inner="quartile", linewidth=1.8, alpha=0.8,
    dodge=False, legend=False, cut=0, ax=ax,
)
sns.boxplot(
    data=normative_train2, x="site_mri_label", y=AGE_COL,
    order=site_order, width=0.15, color="white", linewidth=2.5,
    showfliers=False, boxprops={"zorder": 2}, ax=ax,
)

sm = ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = fig.colorbar(sm, ax=ax, pad=0.03)
cbar.set_label("Sample Size (N)", fontsize=LABEL_SIZE, labelpad=20, fontweight="bold")
cbar.ax.tick_params(labelsize=TICK_SIZE)

ax.set_ylim(8.4, 17)
ax.set_xmargin(0.08)
ax.set_title("Age Distribution Across ABCD Study Sites", fontsize=TITLE_SIZE, fontweight="bold", pad=60)
ax.set_xlabel("MRI Collection Site", fontsize=LABEL_SIZE, labelpad=30, fontweight="bold")
ax.set_ylabel("Age (Years)", fontsize=LABEL_SIZE, labelpad=25, fontweight="bold")
plt.xticks(rotation=45, ha="right", fontsize=TICK_SIZE, fontweight="bold")
plt.yticks(fontsize=TICK_SIZE)
sns.despine()
plt.tight_layout()

path = FIGURES_DIR / "site_age_distribution.png"
plt.savefig(path, bbox_inches="tight")
plt.close()
print(f"Saved: {path}")
````

## Plot 2 — Age distribution: Typical vs Clinical

Grouped violin comparing healthy test vs clinical test  
across the four ABCD visit waves.

````python
healthy_test_f["group"]  = "Typical"
clinical_test_f["group"] = "Atypical"
combined = pd.concat([healthy_test_f, clinical_test_f], axis=0)

palette_grp   = {"Atypical": cc.glasbey_category10[0],
                 "Typical":  cc.glasbey_category10[1]}
sessions_order = ["ses-00A", "ses-02A", "ses-04A", "ses-06A"]

fig, ax = plt.subplots(figsize=(16, 10))
sns.set_style("whitegrid", {"axes.grid": False})

sns.violinplot(
    data=combined, x="session_id", y=AGE_COL, hue="group",
    order=sessions_order, palette=palette_grp,
    dodge=True, inner="quartile", linewidth=1.8, alpha=0.8, cut=0, ax=ax,
)

ax.set_ylim(8.4, 18.5)
ax.set_xlabel("Visit", fontsize=LABEL_SIZE, labelpad=20)
ax.set_ylabel("Age (years)", fontsize=LABEL_SIZE, labelpad=20)
ax.set_title("Age Distribution per Visit: Typical vs Atypical", fontsize=TITLE_SIZE,
             fontweight="bold", pad=50)
ax.legend(title="Group", title_fontsize=22, fontsize=20,
          loc="lower right", bbox_to_anchor=(1, 0))
plt.xticks(fontsize=TICK_SIZE)
plt.yticks(fontsize=TICK_SIZE)
sns.despine()
plt.tight_layout()

path = FIGURES_DIR / "typical_vs_atypical_age.png"
plt.savefig(path, bbox_inches="tight")
plt.close()
print(f"Saved: {path}")
````

## Plot 3 — Individual longitudinal trajectories

One line per participant across the four visit waves.  
Plotted separately for typical (healthy test) and atypical (clinical) groups.

````python
TIME_ORDER = ["00A", "02A", "04A", "06A"]

TRAJ_RC = {
    "font.family": "Arial", "axes.labelsize": 28, "axes.titlesize": 28,
    "xtick.labelsize": 18, "ytick.labelsize": 18, "axes.linewidth": 2,
}

def plot_trajectories(df, title, filename):
    df = df.copy()
    df["event_short"] = df["session_id"].str[-3:]
    df["event_short"] = pd.Categorical(df["event_short"], categories=TIME_ORDER, ordered=True)
    df = df.sort_values(["participant_id", "event_short"])

    plt.rcParams.update(TRAJ_RC)
    sns.set_theme(style="whitegrid")
    sns.set_style({"grid.color": "0.8", "grid.linestyle": "-"})

    fig, ax = plt.subplots(figsize=(9, 6))
    sns.lineplot(
        data=df, x="event_short", y=AGE_COL,
        hue="participant_id", marker="o",
        linewidth=1.8, alpha=0.4, legend=False, palette="tab10", ax=ax,
    )

    ax.set_ylim(8, 19)
    ax.set_yticks(np.arange(8, 19, 1))
    ax.set_xlabel("Timepoint", fontsize=28, labelpad=10)
    ax.set_ylabel("Age (years)", fontsize=28, labelpad=10)
    ax.set_title(title, fontsize=28, pad=15)

    for spine in ax.spines.values():
        spine.set_edgecolor("black")
        spine.set_linewidth(2)
    ax.tick_params(axis="both", width=1.5, color="black")
    plt.grid(alpha=0.5)
    plt.tight_layout()

    path = FIGURES_DIR / filename
    plt.savefig(path, dpi=300, bbox_inches="tight")
    plt.close()
    print(f"Saved: {path}")

plot_trajectories(healthy_test_f,  "Typical Test Sample",   "trajectories_typical.png")
plot_trajectories(clinical_test_f, "Atypical Test Sample",  "trajectories_atypical.png")
````
